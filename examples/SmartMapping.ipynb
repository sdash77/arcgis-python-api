```python
from arcgis.gis import *
from arcgis.geocoding import geocode
```

```python
gis = GIS()
map1 = gis.map('Seattle, WA', zoomlevel=8)
map1
```

```python
from arcgis.features import FeatureLayer
```

```python
fl = FeatureLayer("https://sampleserver6.arcgisonline.com/arcgis/rest/services/Census/MapServer/2")
```

```python
map1.add_layer(fl)
```

```python
map1.add_layer(fl, {'renderer':'ClassedColorRenderer', 'field_name':'POP2007'})
```

```python
map1.add_layer(fl, {"renderer":"ClassedColorRenderer",
               "field_name":"POP2007",
               "opacity":0.75
              })
```

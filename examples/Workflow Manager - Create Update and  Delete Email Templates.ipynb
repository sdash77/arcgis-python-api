# Creating and updating email templates #

This sample will create a new Workflow item and show how to create a email template in workflow manager. 

### Make connections ###

```python
import arcgis
import re
import datetime
from arcgis.gis.workflowmanager import WorkflowManager

# gis = arcgis.gis.GIS(url='https://organizationUrl/portal', username='admin', password='...')
gis = arcgis.gis.GIS(url='https://ps0022953.esri.com//portal', username='admin', password='esri.agp')
item = gis.content.search('title:"Python Sample"')[0]
workflowManager = WorkflowManager(item)
print("Created connection to workflow manager item")
```

### Create a Template ###

```python
# Create email template 
template = {
    "template_name": "Email Template",
    "template_id": "Ef42tu_QQMS-IgZc7pOPnQ",
    "template_details": '{"to":["user@esri.com"],'
    + '"cc":["boss@esri.com"],'
    + '"bcc":["supervisor@esri.com"],'
    + '"subject":"Workflow Manager Templates",'
    + '"body":"Look how easy it is to make a email template!",'
    + '"attachmentSelection":"None",'
    + '"attachmentFolder":null }',
}

return_obj = workflowManager.create_template(template_type="email", 
                                              template_name="Test Email Template",
                                              template_details=template["template_details"],
                                              template_id="Ef42tu_QQMS-IgZc7pOPnQ" )
print(return_obj)

```

### Get the email template and list of templates

```python
my_template = workflowManager.get_template('email', 'Ef42tu_QQMS-IgZc7pOPnQ')

print("My Template")
print(vars(my_template))

templates = workflowManager.templates('email')
print('\n All Templates in my workflow item')
print(templates)
```

###  Update the email template

```python
print(my_template.template_name)

workflowManager.update_template('email', 'Ef42tu_QQMS-IgZc7pOPnQ', 'My New Template Name', my_template.template_details)

updated_template = workflowManager.get_template('email', 'Ef42tu_QQMS-IgZc7pOPnQ')
print(f'The new name for this template is: {updated_template.template_name}')
```

### Delete the email template

```python
print(workflowManager.templates('email'))
return_val = workflowManager.delete_template('email', 'Ef42tu_QQMS-IgZc7pOPnQ')
print(return_val)

print(workflowManager.templates('email'))
```

# Creating and updating email templates #

This sample will create a new workflow item and show how to create an email template in Workflow Manager. 

### Make connections ###

```python
import arcgis
import json
from arcgis.gis.workflowmanager import WorkflowManager

gis = arcgis.gis.GIS(url='https://organizationUrl/portal', username='admin', password='...')

# Search for an existing Workflow Manager Item
item = gis.content.search('title:"Python Example"')[0]
workflowManager = WorkflowManager(item)
print("Created connection to workflow manager item")
```

### Create a Template ###

```python
# Create email template 
details = { "to":["user@esri.com"],
            "cc":["boss@esri.com"],
            "bcc":["supervisor@esri.com"],
            "subject":"Workflow Manager Templates",
            "body":"Look how easy it is to make an email template!",
            "attachmentSelection":"None",
            "attachmentFolder": "" }

return_obj = workflowManager.create_template(template_type="email", 
                                              template_name="Test Email Template",
                                              template_details=details,
                                              template_id="Af42tu_QQMS-IgZc7pOPnQ" )
print(return_obj)

```

### Get the list of templates

```python
# Get the list of email templates
templates = workflowManager.templates('email')
print('All email templates...')
for t in templates:
    print(vars(t))

my_template = workflowManager.get_template('email', 'Af42tu_QQMS-IgZc7pOPnQ')

print("\n My Template")
print(vars(my_template))
```

###  Update the email template

```python
print(my_template.template_name)

workflowManager.update_template('email', 'Af42tu_QQMS-IgZc7pOPnQ', 'My New Template Name', my_template.template_details)

updated_template = workflowManager.get_template('email', 'Af42tu_QQMS-IgZc7pOPnQ')
print(f'The new name for this template is: {updated_template.template_name}')
```

### Delete the email template

```python
print(workflowManager.templates('email'))
success = workflowManager.delete_template('email', 'Af42tu_QQMS-IgZc7pOPnQ')
print(success)

print(workflowManager.templates('email'))
```

# Using Popup Class

## Import Classes

```python
from arcgis.map import Map
from arcgis.map import popups
from arcgis.gis import GIS

gis = GIS(profile="your_online_profile")
```

## Popups
#### Popups are associated with individual layers on a Map. Thus to make edits we can use the Popup class by calling the `popup` method and specifying which layer we will edit

## Editing a Popup Dataclass

```python
# Let's get a layer about current nature reserves in England
feature_layer = gis.content.get("4c496ee272f94e649b32486603bc5e01").layers[0]
feature_layer
```

```python
m2 = Map()
m2.content.add(feature_layer)
m2.content.layers
```

```python
# Let's display the map and see a popup on the layer (Click on a symbol to see this)
m2
```

```python
m2.zoom_to_layer(feature_layer)
```

```python
popup_reserve = m2.content.popup(0)
popup_reserve.info
```

# Creating a Map
**Introduction:**
In this section, we will explore the foundational steps to create a rendered map using the new Map class in Jupyter Lab. The Map class provides a powerful and interactive mapping widget, making it easy to visualize geographical data within your Jupyter environment.

## Import the necessary classes

```python
from arcgis.gis import GIS
from arcgis.map import Map
```

## Create a new Map instance
To create a new map, start by instantiating the Map class. Here's a simple example:

```python
m = Map()
m
```

This basic example sets up a default map, but there's more you can do to customize it. Let's dive into the next section to explore additional options.

## Create a new Map with a specific location
Tailoring your map to a specific location is crucial for focusing on relevant geographical areas. In this section, we'll guide you through specifying the location parameter to create a map with an extent centered around a chosen location.

```python
m2 = Map(location="California")
m2
```

#### The Map class comes equipped with various properties

Use the + / - buttons on the rendered Map and run the cell below to see the zoom value be updated

```python
m2.zoom
```

```python
m2.zoom = 4
```

Pan on the rendered Map and run the cell below to see the extent value be updated

```python
m2.extent
```

```python
m2.extent = {
    "spatialReference": {"latestWkid": 3857, "wkid": 102100},
    "xmin": -20473188.86127435,
    "ymin": 1906235.6873989855,
    "xmax": -6081013.679518911,
    "ymax": 6798205.497648966,
}
```

## Accessing Widgets
Extend the functionality of your map by incorporating various widgets. This section will cover how to access and utilize classes that enable additional features like bookmarks or legends, enhancing the overall user experience.

#### Create a Map from an existing Item

```python
gis = GIS()
webmap_items = gis.content.search("USA Major Cities, owner:esri", item_type="Web Map")
webmap_items
```

```python
m3 = Map(item=webmap_items[0], location="Texas")
m3
```

#### After running the cells below, notice the widgets appear in the top right of the rendered Map

```python
bookmark_manager = m3.bookmarks
bookmark_manager.enabled = True
```

```python
legend_manager = m3.legend
legend_manager.enabled = True
```

```python
layer_list_manager = m3.layer_list
layer_list_manager.enabled = True
```

```python
# If you have a time-enabled layer you can also turn on the time slider
time_slider = m3.time_slider
time_slider.enabled = True
```

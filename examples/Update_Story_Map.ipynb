```python
import os
from arcgis.gis import GIS
from pprint import pprint
```

```python
username = os.environ.get('EGIS_USERNAME')
password = os.environ.get('EGIS_PASSWORD')


# connect to Portal
gis = GIS("https://nwcal-gis-web.nwc.nws.noaa.gov/portal", username, password)

item_id = '55325a63d4b2455f87cd094443d44e53' #change map/app/layer id here

# access the web map
web_service = gis.content.get(item_id)

data = web_service.get_data()
pprint(data)
```

```python
#print the original text in 'description' section
pprint(data['values']['story']['entries'][0]['description']) 

#change the text in the 'description' section
data['values']['story']['entries'][0]['description'] = '<p>It worked!</p>\n'

#print the new text in 'description' section to confirm the text was updated in memory
pprint(data['values']['story']['entries'][0]['description'])
```

```python
pprint(data)
```

```python
#specifiy where to update the text in the service
import json

data_to_update = {'text': json.dumps(data)}
pprint(data_to_update)
```

```python
#perform the update
web_service.update(item_properties=data_to_update)

#get the content again to confirm the update worked
web_service = gis.content.get(item_id)
data = web_service.get_data()

#print the data to make sure the update worked
pprint(data) 
```

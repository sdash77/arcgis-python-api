- Anonymous
- PKI 
- Built-in
- IWA
- LDAP
- Federated Server Login

#### Server Object Examples:

1). Anonymous

```python
from arcgis.server import Server
server = Server(
    url=r"https://sampleserver6.arcgisonline.com/arcgis/rest/services")
```

2). Built-in

```python
server = Server(
    url=r"https://myserver.com/arcgis/rest/services",
    username="user123",
    password="sercret.password1")
```

### Accessing Server from Portal

Besides working with stand alone ArcGIS Server instances.  The Server API support establishing connections to [federated server](http://server.arcgis.com/en/portal/latest/administer/linux/federate-an-arcgis-server-site-with-your-portal.htm).  This requires a user to login to the Portal or ArcGIS Online site then access the hosting and other registered servers through the GIS' object property **servers**.


```python
from arcgis.gis import GIS
portal = GIS('https://myportal.arcgis.com', 
             'user123', 'secret.password1')
servers = portal.servers
```

### Catalog View

The catalog view of server is the non-administrative side of server.  With the catalog view, users can access feature layers, map services, geoprocessing tool, and much more to perform analysis or a variety of other tasks.  The catalog only shows services by folder.  Once a user changes the folder, they can access services in that current space.


```python
server = Server(
    url=r"https://sampleserver1.arcgisonline.com/arcgis/rest/services")
catalog = server.catalog
```

Now that an anaonymous catalog view has been created, a user can set the folder they want to work with and start using content. 

```python
catalog.folder = 'Demographics'
for service in catalog.services:
    print(service)
```

### Working with Services

Services can be easily accessed by either accessing them directly or walking the server's catalog.


#### Access Services from Catalog

The services property on the catalog object, like all Python lists, is zero based.  This means the first entry is 0, not 1.  1 is the second entry in the list.  

```python
mil = catalog.services[0]
query_results = mil.layers[1].query(where='1=1', return_all_records=False) # gets first 1000 records only
```

#### Access Service Directly from URL

Services can also be leveraged directly from the URL.

```python
from arcgis.server import Service

url = "https://sampleserver1.arcgisonline.com/arcgis/rest/services/Demographics/ESRI_Census_USA/MapServer/1"
service = Service(url_or_item=url)
query_results = service.query(where='1=1', return_all_records=False) # gets first 1000 records only
```

### Administration of Server

If an administrator uses the Python API for Arcgis, they can administer the ArcGIS Server by accessing the properties off of the Server object.  

- **data** - accesses the datastore on server
- **info** - A read-only resource that returns meta information about the server
- **kml** - This resource is a container for all the KMZ files created on the server
- **logs** - Logs are the records written by the various components of ArcGIS Server. You can query the logs and change various log settings
- **machines** - This resource represents a collection of all the server machines that have been registered with the site
- **services** - This resource represents the administation side of managing services and folders on ArcGIS Server
- **system** - The System resource is a collection of miscellaneous server-wide resources
- **usage** - This resource is a collection of all the usage reports created within your site
- **users** - used to manage and modify users on a non-federated server

```python
from arcgis.server import Server
server = Server(
    url="http://myserver.com/arcgis",
username='admin',
password="sercret.password1")
logs = server.logs
query_results = logs.query()
```

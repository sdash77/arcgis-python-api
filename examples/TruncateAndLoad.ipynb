```python
from arcgis.gis import *
from arcgis.tools import *
from arcgis.lyr import *
```

```python
gis = GIS("OrganzationURL", "UserName", "Password")
svc = FeatureService("Url to Feature Service", gis)
```

```python
layer = svc.layers
print ("Service contains {0} layers".format(len(layer)))
```

```python
fl=layer[0]
print ("First layers url is: {0}".format(fl.url))
```

```python
oidQResults = fl.query(returnIDsOnly=True)
if len(oidQResults['objectIds']) > 0:
    print ("Service contain {0} features".format(len(oidQResults['objectIds'])))
    oidString = ','.join(str(x) for x in oidQResults['objectIds'])
    fl.edit_features(deletes=oidString)
else:
    print ("Service does not contain any features")
```

```python
feat = Feature()
features = feat.fc_to_features(r'Path to GDB Layer')
```

```python
featDict = [feat.as_dict for feat in features]
```

```python
if len(features) > 0:
    addresult = fl.edit_features(adds=featDict)
    print (addresult)
else:
    print ("The layer does not contain any features")
```

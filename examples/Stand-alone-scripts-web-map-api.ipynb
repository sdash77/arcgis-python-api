# Using web map API in stand alone scripts

```python
from arcgis.gis import *
from arcgis.layers import WebMap
```

```python
gis = GIS("https://www.arcgis.com","arcgis_python")
# anon_gis= GIS()
```

## Create empty web map object

```python
wm = WebMap()
wm.definition
```

# Adding different layer types

## Feature Layer item

```python
flc_item = gis.content.get('8d29d4a521f34ebba40dd918ac8322c3')
flc_item
```

```python
wm.add_layer(flc_item)
```

```python
wm.definition.operationalLayers
```

```python
wm._extent
```

```python
wm_item = wm.save_as(title='WM standalone FLC item', snippet='automation',tags=['automation'])
wm_item
```

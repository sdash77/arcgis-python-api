```python
from datetime import datetime
```

```python
from arcgis.gis import GIS
gis = GIS("https://deldev.maps.arcgis.com", "demos_deldev", "DelDevs12")
```

```python
origin_item = gis.content.get('1f163950afbb4bf39f3ee67cf411761d')
dest_item = gis.content.get('c7665d3c8e6f48a79f07b79677996bed')
point_barrier = gis.content.get('d692355520e94d39a028f79248b75ef7')
```

```python
origin_layer = origin_item.layers[0]
dest_layer = dest_item.layers[0]
barrier_lyr = point_barrier.layers[0]
```

```python
from arcgis.features.use_proximity import connect_origins_to_destinations
```

```python
route1 =  connect_origins_to_destinations(origins_layer=origin_layer,
                                         destinations_layer=dest_layer,
                                         measurement_type='DrivingTime',
                                         time_of_day=631371600000,
                                         output_name="test_route1" + str(datetime.now().microsecond)
                                        )
```

```python
route1.delete()
```

```python
route2 =  connect_origins_to_destinations(origins_layer=origin_layer,
                                         destinations_layer=dest_layer,
                                         measurement_type='StraightLine', # line distance
                                         time_of_day=631371600000,
                                         output_name="test_route2" + str(datetime.now().microsecond)
                                        )
```

```python
route2.delete()
```

```python
import arcgis.network as network
route_service = network.RouteLayer(gis.properties.helperServices.route.url, gis=gis)
print(sorted([i['name'] for i in route_service.retrieve_travel_modes()['supportedTravelModes']]))
```

```python
travel_mode = [i for i in route_service.retrieve_travel_modes()['supportedTravelModes'] 
            if i['name'] == 'Rural Driving Distance'][0]
```

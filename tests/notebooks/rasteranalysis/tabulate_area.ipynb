```python
from datetime import datetime
```

```python
import arcgis
from arcgis.gis import GIS

gis = GIS("https://sha-97611-d02.esri.com/portal", "", "", verify_cert=False)
gis
```

```python
arcgis.env.verbose=True
```

```python
degs=gis.content.search("title: degs")[0].layers[0]
degs.url
```

```python
negs=gis.content.search("title: negs")[0].layers[0]
negs.url
```

```python
from arcgis.raster.analytics import tabulate_area
```

```python
tabulate_area_op = tabulate_area(input_zone_raster_or_features=degs, 
                                 zone_field="VALUE",
                                 input_class_raster_or_features=degs,
                                 class_field= "VALUE",
                                 output_name="tabulate_area" + str(datetime.now().strftime('%Y%m%d%H%M%S')),
                                 gis=gis)
tabulate_area_op
```

```python
tabulate_area_op = tabulate_area(input_zone_raster_or_features=degs, 
                                 zone_field="VALUE",
                                 input_class_raster_or_features=degs,
                                 class_field= "VALUE",
                                 classes_as_rows=False,
                                 analysis_cell_size=10,
                                 output_name="tabulate_area" + str(datetime.now().strftime('%Y%m%d%H%M%S')),
                                 gis=gis)
tabulate_area_op   
```

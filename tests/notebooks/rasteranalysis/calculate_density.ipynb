```python
from datetime import datetime
```

```python
import arcgis
from arcgis.gis import GIS

gis = GIS("https://pythonapi.playground.esri.com/portal", "arcgis_python", "amazing_arcgis_123")
gis
```

```python
arcgis.env.verbose=True
```

```python
sf_custs_shp=gis.content.search("sf_custs_shp", item_type="Feature Layer Collection")[0].layers[0]
sf_custs_shp
```

Calculate Density with arcgis.env.analysis_extent param set for synchronous job

```python
aoi={"xmin": -13633946.5247196, "ymin": 4544946.0757069, "xmax": -13624287.111718, "ymax": 4552826.60826073, "spatialReference": {"wkid": 3857}}
```

```python
calculate_density_1 = arcgis.raster.analytics.calculate_density(input_point_or_line_features=sf_custs_shp,
                                                                count_field="NEAR_FID",
                                                                search_distance=None,
                                                                output_area_units=None,
                                                                output_cell_size=None,
                                                                output_name="calculate_density_" + str(datetime.now().strftime('%Y%m%d%H%M%S')),
                                                                gis=gis)
assert calculate_density_1
```

Calculate density with arcgis.env.analysis_extent param reset for synchronous job

```python
calculate_density_2 = arcgis.raster.analytics.calculate_density(input_point_or_line_features = sf_custs_shp,
                                                                count_field = "SALES",
                                                                search_distance = {"distance":50,"units":"Meters"},
                                                                output_area_units = "Square Meters",
                                                                output_cell_size = {"distance":50,"units":"Meters"},
                                                                output_name = "calculate_density_" + str(datetime.now().strftime('%Y%m%d%H%M%S')),
                                                                context={"extent":aoi},
                                                                gis = gis)
assert calculate_density_2
```

Calculate density using arcgis.env.analysis_extent parameter and excluding 'output_name' parameter for asynchronous job.

```python
arcgis.env.analysis_extent = aoi
calculate_density_3 = arcgis.raster.analytics.calculate_density(input_point_or_line_features = sf_custs_shp,
                                               count_field = "SALES",
                                               search_distance = {"distance":50,"units":"Meter"},
                                               output_area_units = "Square Meters",
                                               output_cell_size = {"distance":50,"units":"Meter"},
                                               future = True,
                                               gis = gis)
assert calculate_density_3.result()
```

```python
import os
from datetime import datetime
```

```python
import arcgis
from arcgis.gis import GIS

gis = GIS("https://pythonapi.playground.esri.com/portal", "arcgis_python", "amazing_arcgis_123")
gis
```

```python
arcgis.env.verbose=True
```

Get Registered Datastores

```python
ds=arcgis.raster.analytics.get_datastores(gis=gis)
```

```python
fileshare_1=ds.get("/fileShares/ListDatastoreContent")
```

List datastore content for datastore object item

```python
list_ds_content_1 = arcgis.raster.analytics.list_datastore_content(datastore=fileshare_1,
                                                                   future=True)
list_ds_content_1.result()
```

List datastore content using fileshare path

```python
list_ds_content_3 = arcgis.raster.analytics.list_datastore_content(datastore="/fileShares/ListDatastoreContent/files")
list_ds_content_3
```

List .xml files in fileshare path

```python
list_ds_content_4 = arcgis.raster.analytics.list_datastore_content(datastore="/fileShares/ListDatastoreContent/files", 
                                                                   filter="*.xml",
                                                                   gis=gis,
                                                                   future=True)
list_ds_content_4.result()
```

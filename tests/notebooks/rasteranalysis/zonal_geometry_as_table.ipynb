```python
from datetime import datetime
```

```python
import arcgis
from arcgis.gis import GIS

gis = GIS("https://sha-97611-d02.esri.com/portal", "", "", verify_cert=False)
gis
```

```python
arcgis.env.verbose=True
```

```python
degs=gis.content.search("title: degs")[0].layers[0]
degs.url
```

```python
negs=gis.content.search("title: negs")[0].layers[0]
negs.url
```

```python
from arcgis.raster.analytics import zonal_geometry_as_table
```

```python
zonal_geometry_as_table_op = zonal_geometry_as_table(input_zone_raster_or_features=degs, 
                                                     zone_field="VALUE", 
                                                     output_name="zonal_geometry_as_table" + str(datetime.now().strftime('%Y%m%d%H%M%S')),
                                                     gis=gis)
zonal_geometry_as_table_op
```

```python
zonal_geometry_as_table_op = zonal_geometry_as_table(input_zone_raster_or_features=degs, 
                                                     zone_field="VALUE", 
                                                     analysis_cell_size=10,
                                                     output_name="tabulate_area" + str(datetime.now().strftime('%Y%m%d%H%M%S')),
                                                     gis=gis)
zonal_geometry_as_table_op
```

```python
from datetime import datetime
```

```python
import arcgis
from arcgis.gis import GIS

#gis = GIS("https://pythonapi.playground.esri.com/portal", "arcgis_python", "amazing_arcgis_123")
profile = 'your_enterprise_profile'
gis = GIS(profile=profile, verify_cert=False)
gis
```

```python
arcgis.env.verbose=True
```

```python
geonamesras=gis.content.search("geonamesras")[0].layers[0]
geonamesras.url
```

```python
cd20=gis.content.search("cd20", item_type="Image Service")[0].layers[0]
cd20.url
```

```python
bl20=gis.content.search("bl20")[0].layers[0]
bl20.url
```

Cost path as polyline basic test case

```python
from datetime import datetime
costpath_as_polyline_1 = arcgis.raster.analytics.determine_travel_costpath_as_polyline(input_source_data=geonamesras,
                                                                                       input_cost_raster=cd20,
                                                                                       input_destination_data=bl20,
                                                                                       output_polyline_name="costpath_as_polyline"+str(datetime.now().strftime('%Y%m%d%H%M%S')),
                                                                                       destination_field=None,
                                                                                       gis=gis,
                                                                                       folder="costpath_as_polyline")
assert costpath_as_polyline_1
```

Cost path as polyline run asynchronously

```python
costpath_as_polyline_2 = arcgis.raster.analytics.determine_travel_costpath_as_polyline(input_source_data=geonamesras,
                                                                                       input_cost_raster=cd20,
                                                                                       input_destination_data=bl20,
                                                                                       output_polyline_name="costpath_as_polyline"+str(datetime.now().strftime('%Y%m%d%H%M%S')),
                                                                                       destination_field=None,
                                                                                       gis=gis,
                                                                                       future=True)
assert costpath_as_polyline_2.result()
```

```python
from datetime import datetime
```

```python
import arcgis
from arcgis.gis import GIS

#gis = GIS("https://sha-imgcl-d01.esri.com/portal", "admin", "adminadmin")

profile = 'your_enterprise_profile'
gis = GIS(profile=profile, verify_cert=False)
gis
```

```python
arcgis.env.verbose=True
```

```python
elevcond8 = gis.content.search("title:elevcond8")[0].layers[0]
elevcond8.url
```

```python
zone_int = gis.content.search("title:zone_int")[0].layers[0]
zone_int.url
```

Convert raster to line features using to_features()

```python
convert_raster_to_feature_1 = elevcond8.to_features(field='Value', output_type='Line', simplify=False, gis=gis)
convert_raster_to_feature_1
```

Convert raster to polygon features using to_features()

```python
convert_raster_to_feature_2 = zone_int.to_features(future=True)
convert_raster_to_feature_2.result()
```

Set context param for convert raster to polygon feature.

```python
aoi = {"xmin": 483443.37822144,"ymin": 208728.88678416,"xmax": 494094.61662144,"ymax": 219724.680012852,"spatialReference": {"wkt": "PROJCS[\"NAD_1983_Transverse_Mercator\",GEOGCS[\"GCS_North_American_1983\",DATUM[\"D_North_American_1983\",SPHEROID[\"GRS_1980\",6378137.0,298.257222101]],PRIMEM[\"Greenwich\",0.0],UNIT[\"Degree\",0.0174532925199433]],PROJECTION[\"Transverse_Mercator\"],PARAMETER[\"False_Easting\",500000.0],PARAMETER[\"False_Northing\",0.0],PARAMETER[\"Central_Meridian\",-72.5],PARAMETER[\"Scale_Factor\",0.999964286],PARAMETER[\"Latitude_Of_Origin\",42.5],UNIT[\"Meter\",1.0]]"}}
```

```python
from datetime import datetime
convert_raster_to_feature_3 = arcgis.raster.analytics.convert_raster_to_feature(input_raster=zone_int,
                                                                                field='Value', 
                                                                                output_type='Polygon',
                                                                                simplify=False,
                                                                                output_name="convert_raster_to_feature_" +str(datetime.now().strftime('%Y%m%d%H%M%S')),
                                                                                context={"extent":aoi},
                                                                                future=False,
                                                                                gis=gis)
convert_raster_to_feature_3
```

Set environment variable analysis_extent for convert raster to polygon Feature

```python
arcgis.env.analysis_extent = aoi
convert_raster_to_feature_4 = arcgis.raster.analytics.convert_raster_to_feature(input_raster=zone_int,
                                                                                field='Value', 
                                                                                output_type='Polygon',
                                                                                simplify=True,
                                                                                output_name="convert_raster_to_feature_" +str(datetime.now().strftime('%Y%m%d%H%M%S')),
                                                                                future=True,
                                                                                gis=gis)
convert_raster_to_feature_4.result()
```

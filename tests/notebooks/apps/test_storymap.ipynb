```python
# Folder path
content = r"C:\ipython_workfolder\Content"
```

### Create A New Story and Add Items

```python
# Create a new StoryMap
new_story = StoryMap()
new_story.node_order
```

```python
# Create Story Content
canyon = Image(content+r"\canyon.png")
trees = Image(content+r"\trees.jpg")

bear = Video(content+r"\bear.mp4")
fawn = Video(content+r"\fawn.mp4")

craine = Audio(content+r"\craine.mp3")

park_media = WebPage("https://www.nps.gov/media/multimedia-search.htm#sort=Date_Last_Modified+desc")
national_parks = WebPage("https://www.nps.gov/index.htm")

leaves = Button(link = "https://www.nps.gov/subjects/forests/leaf-peeping.htm", text="Autumn Colors")
bats = Button(link = "https://www.nps.gov/subjects/bats/index.htm", text= "Look at Bats")

parks = Map("f092c20803a047cba81fbf1e30eff0b5") #living atlas: USA Parks

welcome = Text(text= "Welcome to a New Story About Some National Park Information", style= "heading")
quote = Text(text="I encourage everybody to hop on Google and type in ‘national park’ in whatever state they live in and see the beauty that lies in their own backyard. It’s that simple.", style="quote")
```

```python
# Edit story cover
new_story.story_cover("My First Story About National Parks", type="minimal", summary="Testing the Python API", by_line="Python Tester")
```

```python
# Add Items to the story
new_story.add(canyon, "Here is an image of a canyon", "A nice canyon", "float")
new_story.add(trees, "Some trees with a deer", "Sequoia trees in the distance", "full")
new_story.add() #separator

new_story.add(bear)
video = new_story.add(fawn)
new_story.add()

new_story.add(craine)
new_story.add()

web_page = new_story.add(park_media, display="small")
new_story.add()

new_story.add(leaves)
new_story.add(bats)
new_story.add()

new_story.add(parks, "A Map of US Parks", display="medium")

new_story.add(welcome, position=2)
new_story.add(position=3)
new_story.add(quote, position=4)
new_story.add(position=5)
```

```python
# See all nodes
new_story.node_order
```

```python
# See all text nodes
new_story.list_nodes("text")
```

```python
# See webpage node before updating 
new_story.properties["nodes"][web_page]
```

```python
# Update webpage node
new_story.update(web_page, national_parks)
new_story.properties["nodes"][web_page]
```

```python
# Move a node
new_story.move(video, position=9)
new_story.node_order
```

```python
# Delete node
new_story.delete(video)
new_story.node_order
```

```python
# Save the entire story
new_story.save(title="National Parks Test")
```

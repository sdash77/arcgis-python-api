### Create A New Story and modify 


```python
my_story.theme(Themes.SLATE)
# Edit story cover
my_story.story_cover(
    "My First Story About National Parks",
    type="minimal",
    summary="Testing the Python API",
    by_line="Python Tester",
)
my_story.save()
```

### Add Items


Create an image from a url and add it to the story with caption and alt_text.  
Save my story again and assign a new title this time.

```python
trees = Image("https://www.nps.gov/npgallery/GetAsset/69680c29-caa3-42da-93d9-32925e9ed409/proxy/hires")
my_story.add(trees, "Some trees with a deer", "Sequoia trees in the distance")
my_story.add()  # separator
my_story.save("My Small National Park Story")
```

```python
trees.properties
```

Create Video and Audio nodes from file paths to add to story.

```python
bear = Video(content + r"\bear.mp4")
video = my_story.add(bear)


craine = Audio(content + r"\craine.mp3")
print("Node Order Before Adding Audio:")
print(my_story.node_order)

audio = my_story.add(craine, position=2)
separator = my_story.add()
print("Node Order After Adding Audio:")
print(my_story.node_order)


my_story.save()
```

Create Embed, Button, Map, Text, Button to add to story

```python
park_media = Embed(
    "https://www.nps.gov/media/multimedia-search.htm#sort=Date_Last_Modified+desc",
    my_story,
)
web_page = my_story.add(park_media)

leaves = Button(
    link="https://www.nps.gov/subjects/forests/leaf-peeping.htm",
    text="Autumn Colors",
    story=my_story,
)
leaf_button = my_story.add(leaves)

covid_map = Map("006c10be294f4f64a5c9e5202cfa64c3", story=my_story)
covid = my_story.add(
    covid_map, caption="This has nothing to do with parks but I needed a map"
)

welcome = Text(
    story=my_story,
    text="Welcome to a New Story About Some National Park Information",
    style=TextStyles.HEADING,
)
my_story.add(welcome, position=2)
quote = Text(
    story=my_story,
    text="I encourage everybody to hop on Google and type in ‘national park’ in whatever state they live in and see the beauty that lies in their own backyard. It’s that simple.",
    style=TextStyles.QUOTE,
)
my_story.add(quote, position=4)

my_story.save()
```

### Adding Basics

Change story theme by using the enumerated Themes class and choosing a type of theme.

### Create Embed, Button, Map, and Texts

Embed can be created with a url.  
Button can be created with a text and a link.  
Map is created from an exisiting map item or item id of type webmap.  
Text is created from text and a style of text.  

```python
# Create Embed from a link
park_media = Embed(
    "https://www.nps.gov/media/multimedia-search.htm#sort=Date_Last_Modified+desc"
)
web_page = my_story.add(park_media)

# Create a Button
leaves = Button(
    link="https://www.nps.gov/subjects/forests/leaf-peeping.htm",
    text="Autumn Colors"
)
leaf_button = my_story.add(leaves)

# Create two types of Text and add to certain position in the story.
welcome = Text(
    text="Welcome to a New Story About Some National Park Information",
    style=TextStyles.HEADING,
)
my_story.add(welcome, position=2)
quote = Text(
    text="I encourage everybody to hop on Google and type in ‘national park’ in whatever state they live in and see the beauty that lies in their own backyard. It’s that simple.",
    style=TextStyles.QUOTE,
)
my_story.add(quote, position=4)
```

```python
# Create a Map from an exisiting webmap
my_map = Map("787722665db94bfa80fe36bd38848430")
my_story.add(
    my_map, caption="This has nothing to do with parks but I needed a map"
)
```

## Edit an Existing Story

To recreate this sample story:  
1. Create a new story
2. Add what you want to the story but must have at least one Sidecar
3. In Sidecar have at minimum three slides. The third slide in the sample story can be a map like in the example.
4. Publish the story or save it then copy the item id and replace the one below.

```python
# Pass in item id for an exisitng story
sample_story = StoryMap("0b6b458cea604385ba1398ef8e38a494")
```

```python
# See node order of the sample
sample_story.nodes
```

### Edit a slide of Sidecar

```python
# Create an instance of image
image1 = Image(content + r"\wolves.jpg")
```

```python
# Create an instance of Sidecar from an exisitng node
sidecar = sample_story.content_list[5]
```

```python
# Properties shows each slide and their content nodes.
sidecar.properties
```

```python
# Edit image on the fourth slide to the new image created above
sidecar.edit(image1, 4)
```

```python
# Notice the third media item node has now changed.
sidecar.properties
```

```python
sample_story.save()
```

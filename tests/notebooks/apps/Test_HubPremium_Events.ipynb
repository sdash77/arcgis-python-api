```python
from arcgis.gis import GIS
```

```python
gis_premium = GIS("https://www.arcgis.com", "hubpy_test", "hubPython01")
myhub = gis_premium.hub
```

```python
#test get
e = myhub.events.get(1)
assert e.title=='Test'
```

```python
#test add
event_properties = {
    "title":"Event created using python 2022",
    'description': 'Testing with python',
    'initiativeId': 'b4cbdaf65f3c4b55a8fb1f8dcbe0fc3b',
    'venue': 'ESRI R&D Center',
    'address1': '1001 19th St N #910, Arlington, VA 22209',
    'status': 'public',
    'startDate': 1585356198,
    'endDate': 1585399398,
    'isAllDay': 0,
    'geometry': {'x': -77.06915099999999, 'y': 38.89740599999999}
}

new_event = myhub.events.add(event_properties)
assert new_event.title==event_properties['title']
```

```python
#test search
searched = myhub.events.search(title=event_properties['title'])
assert searched[0].event_id==new_event.event_id
```

```python
#test update
new_event.update(event_properties={'status':'private'})
assert new_event.access=='private'
```

```python
def test_events_delete(event_id):
    '''
    Tests if event is successfully deleted
    '''
    try:
        return myhub.events.get(event_id)
    except:
        return 'Works as expected'
```

```python
#test delete
new_event.delete()
assert test_events_delete(new_event.event_id)=='Works as expected', 'Still Exists'
```

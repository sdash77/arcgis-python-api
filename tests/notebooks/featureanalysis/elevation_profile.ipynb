```python
from arcgis.gis import GIS
from arcgis.features import FeatureCollection, FeatureSet
from arcgis.gis import Item
from arcgis.geoprocessing._job import GPJob
gis = GIS('https://geosaurus.maps.arcgis.com', 'arcgis_python', 'P@ssword123')
```

```python
import arcgis.features.elevation
```

```python
from arcgis.features.elevation import profile
```

```python
item = gis.content.get('3428672decd64aeabcb8da15e9b749f0')
```

```python
item
```

```python
lyr = item.layers[0]
```

```python
fs = lyr.query(where='METERS=14023')
```

```python
import arcgis.env
```

```python
import arcgis
```

```python
arcgis.env.return_m = True
```

```python
arcgis.env.return_z = True
```

```python
result = profile(input_line_features=fs,
                 dem_resolution='FINEST',
                 #maximum_sample_distance=1000,
                 maximum_sample_distance_units='Meters', future=True)
```

```python
assert isinstance(result, GPJob)
assert isinstance(result.result(), FeatureSet)
```

```python
res = result.result()
res
```

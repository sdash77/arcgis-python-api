```python
#import sys
#sys.path.insert(0, r"C:\\ipython_workfolder\\geosaurus\\src")
from arcgis.gis import GIS

portal = GIS(profile="your_portal_profile", verify_cert=False,)
```

```python
airport_item = portal.content.get("8c3a85d4032049ff8333965c6a2a6219")
airport_layer = airport_item.layers[2]
seaport_layer = airport_item.layers[5]
```

```python
test= find_point_clusters(airport_layer,
                           min_features_cluster=200,
                           search_distance=2,
                           search_distance_unit='Kilometers',
                           output_name='test point clusters overwrite')
```

```python
fl_to_overwrite = portal.content.get(test.id).layers[0]
print(fl_to_overwrite)
print(test.id)
```

```python
overwrite= find_point_clusters(seaport_layer,
                           min_features_cluster=200,
                           search_distance=2,
                           search_distance_unit='Kilometers',
                           output_name=fl_to_overwrite)
```

```python
# Notice the ids are the same, if overwrite fails then fl is appended instead
print(overwrite.layers)
print(overwrite.id)
```

```python
overwrite.delete()
```

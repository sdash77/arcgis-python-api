```python
from datetime import datetime
from arcgis.features import FeatureCollection, FeatureSet
from arcgis.gis import Item
from arcgis.geoprocessing._job import GPJob
```

```python
from arcgis.gis import GIS
#gis = GIS(profile="your_online_profile")
gis = GIS('http://deldev.maps.arcgis.com', 'demos_deldev', 'DelDevs12')
```

```python
item = gis.content.get("79d3e458dcaf486f81c5591a67538179")
```

```python
item
```

```python
collision = item.layers[0]
```

```python
from arcgis.features.analyze_patterns import find_point_clusters
```

```python
from datetime import datetime
test = find_point_clusters(collision, min_features_cluster=50, 
                           output_name='test point clusters' + str(datetime.now().microsecond))
```

```python
assert isinstance(test, Item)
test
```

```python
test.delete()
```

```python
test2= find_point_clusters(collision,
                           min_features_cluster=200,
                           search_distance=2,
                           search_distance_unit='Kilometers',
                           output_name='test point clusters' + str(datetime.now().microsecond))
```

```python
test2
```

```python
test2.delete()
```

```python
test4= find_point_clusters(collision,
                           min_features_cluster=200,
                           search_distance=2,
                           future=True,
                           search_distance_unit='Kilometers')
assert isinstance(test4, GPJob)
```

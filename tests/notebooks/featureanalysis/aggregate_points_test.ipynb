```python
def id_generator(size=6, chars=string.ascii_uppercase + string.digits):
    return ''.join(random.choice(chars) for _ in range(size))
```

#### Test Overwrite

```python
#import sys
#sys.path.insert(0, r"C:\\ipython_workfolder\\geosaurus\\src")
from arcgis.gis import GIS
from arcgis.features.analysis import aggregate_points

portal = GIS(profile="your_portal_profile", verify_cert=False) #need to be at least enetrprise 10.9.1
```

```python
# Commercial permits since 2010 FL
point_item = portal.content.get("8e4e67a06ff247a9870427f19e7d744c")
# US zip code areas FL
polygon_item = portal.content.get("d6f7ee6129e241cc9b6f75978e47128b")

point_layer = point_item.layers[0]
polygon_layer = polygon_item.layers[0]
```

```python
# Feature Layer to overwrite
fl_to_overwrite = aggregate_points(
    point_layer=point_layer,
    polygon_layer=polygon_layer,
    keep_boundaries_with_no_points=False,
    output_name="testing_agg_overwrite",
)
```

```python
# feature layer to be overwritten
fl_id = fl_to_overwrite.id
fl_id
```

```python
fl = portal.content.get(fl_id).layers[0]
fl
```

```python
# Perform overwrite
aggregate_overwrite = aggregate_points(
    point_layer=point_layer,
    polygon_layer=polygon_layer,
    keep_boundaries_with_no_points=False,
    output_name=fl,
    context={"overwrite": True},
)
```

```python
# Notice item ids are the same, successfully overwritten
aggregate_overwrite.id
```

```python
fl
```

```python
aggregate_overwrite.delete()
```

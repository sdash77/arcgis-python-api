```python
import sys
import uuid
from arcgis.gis import GIS, Item
from arcgis.features import FeatureCollection, FeatureSet
from arcgis.gis import Item
from arcgis.geoprocessing._job import GPJob
```

```python
from arcgis.features.manage_data import generate_tessellation
```

```python
gis = GIS(profile="your_online_profile")
#gis = GIS("https://deldev.maps.arcgis.com", "demos_deldev", "DelDevs12")
item = gis.content.get('0e199091d6b9416b9abb6655e1975d44')
layer = item.layers[0]
uid = uuid.uuid4().hex[:5]
```

```python
item
```

```python
tbx = gis._tools.featureanalysis._tbx
```

```python
est = gt1 = generate_tessellation(extent_layer=layer,
                           output_name="thisisatest" +uid,
                           bin_size=100,
                           bin_size_unit="SquareKilometers",
                           bin_type="SQUARE",estimate=True)
assert isinstance(est, dict)
est
```

```python
gpjob = generate_tessellation(extent_layer=layer,
                              output_name="thisisatest" +uid,
                              bin_size=100,
                              bin_size_unit="SquareKilometers",
                              bin_type="SQUARE",future=True)
assert isinstance(gpjob, GPJob)
res = gpjob.result()
assert isinstance(res, Item)
res
```

```python
assert res.delete()
```

```python
gt = generate_tessellation(extent_layer=layer,
                           bin_size=100,
                           bin_size_unit="SquareKilometers",
                           bin_type="SQUARE",)
assert isinstance(gt, FeatureCollection)
gt
```

```python
job = generate_tessellation(extent_layer=layer,
                           bin_size=100,
                           bin_size_unit="SquareKilometers",
                           future=True,
                           bin_type="SQUARE",)
assert isinstance(job, GPJob)
gt = job.result()
assert isinstance(gt, FeatureCollection)
gt
```

```python
uid = uuid.uuid4().hex[:5]
gt1 = generate_tessellation(extent_layer=layer,
                           output_name="thisisatest" +uid,
                           bin_size=100,
                           bin_size_unit="SquareKilometers",
                           future=True,
                           bin_type="SQUARE",)
assert isinstance(gt1.result(), Item)
assert gt1.result().delete()
```

```python
test = generate_tessellation(extent_layer=bike_layer,
                              output_name="tesselation_overwrite",
                              bin_size=100,
                              bin_size_unit="SquareKilometers",
                              bin_type="SQUARE")
```

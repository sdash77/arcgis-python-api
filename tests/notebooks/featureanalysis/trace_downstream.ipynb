```python
item = gis.content.get('1f163950afbb4bf39f3ee67cf411761d')
```

```python
item
```

```python
from datetime import datetime
item = trace_downstream(input_layer=item.layers[0],
                 split_distance=2,
                 split_units='Miles',
                 max_distance=2,
                 max_distance_units='Miles',
                 bounding_polygon_layer=None,
                 source_database='Finest',
                 generalize=True,
                 output_name='test downstream' + str(datetime.now().microsecond))
item
```

```python
#import sys
#sys.path.insert(0, r"C:\\ipython_workfolder\\geosaurus\\src")
from arcgis.gis import GIS

portal = GIS(profile="your_portal_profile", verify_cert=False,)
```

```python
#airport item
item = portal.content.get("8c3a85d4032049ff8333965c6a2a6219")
layer1 = item.layers[5]
layer2 = item.layers[3]
```

```python
test = trace_downstream(input_layer=layer1,
                 split_distance=2,
                 split_units='Miles',
                 max_distance=2,
                 max_distance_units='Miles',
                 bounding_polygon_layer=None,
                 source_database='Finest',
                 generalize=True,
                 output_name='test_downstream_overwrite')
```

```python
fl_to_overwrite = portal.content.get(test.id).layers[0]
print(fl_to_overwrite)
print(test.id)
```

```python
overwrite = trace_downstream(input_layer=layer2,
                 split_distance=2,
                 split_units='Miles',
                 max_distance=2,
                 max_distance_units='Miles',
                 bounding_polygon_layer=None,
                 source_database='Finest',
                 generalize=True,
                 output_name=fl_to_overwrite,
                 context={"overwrite":True})
```

```python
# Notice the ids are the same. If overwrite failed then fl is appended instead
print(overwrite.layers)
print(overwrite.id)
```

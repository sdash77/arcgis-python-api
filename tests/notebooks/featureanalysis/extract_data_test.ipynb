```python
from datetime import datetime
from arcgis.features import FeatureCollection, FeatureSet
from arcgis.gis import Item
from arcgis.geoprocessing._job import GPJob
from arcgis.gis import GIS
# gis = GIS(profile="your_online_profile")
```

```python
gis = GIS("https://deldev.maps.arcgis.com", "demos_deldev", "DelDevs12")
```

```python
from arcgis.features.manage_data import extract_data
```

```python
items = gis.content.search('tags:cougar_habitat owner:demos_deldev', 'Feature Layer', outside_org=True)
```

```python
highways = gis.content.get('0707fd77e3bf4f0ab443b71425953152')
```

```python
highways
```

```python
state_area_boundary = gis.content.get('a6cb2a0688d841fd803cd82b4d8282b4')
```

```python
state_area_boundary
```

```python
from datetime import datetime
```

```python
ext_state_highway1 = extract_data(input_layers=[highways.layers[0]],
                                 extent=state_area_boundary.layers[0],
                                 clip=True,
                                 data_format='shapefile',
                                 output_name='piyatesting3' + str(datetime.now().microsecond))
```

```python
assert isinstance(ext_state_highway1, Item)
ext_state_highway1.delete()
```

```python
ext_state_highway1 = extract_data(input_layers=[highways.layers[0]],
                                 extent=state_area_boundary.layers[0],
                                 clip=True,
                                  future=True,
                                 data_format='shapefile',
                                 output_name={ "title": "test_" + str(datetime.now().microsecond) ,
                                               "tag": "datascience",
                                               "snippet": "none",
                                               "description": "extract data"})
```

```python
assert isinstance(ext_state_highway1, GPJob)
assert isinstance(ext_state_highway1.result(), Item)
assert ext_state_highway1.result().delete()
```

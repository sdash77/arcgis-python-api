```python
from datetime import datetime
from arcgis.features import FeatureCollection, FeatureSet
from arcgis.gis import Item
from arcgis.geoprocessing._job import GPJob

from arcgis.gis import GIS
gis = GIS('http://deldev.maps.arcgis.com', 'demos_deldev', 'DelDevs12', verify_cert=False)
```

```python
from arcgis.features.summarize_data import summarize_nearby
```

```python
item1 = gis.content.get("1f163950afbb4bf39f3ee67cf411761d")
```

```python
item2 = gis.content.get('6e22847e65c2449e8d6bc494e8f417e5')
item2
```

```python
item1.layers[0]
```

```python
from datetime import datetime
item = summarize_nearby(sum_nearby_layer=item2.layers[0],
                 summary_layer=item1.layers[0],
                 near_type='StraightLine',
                 distances=[5],
                 units='Miles',
                 time_zone_for_time_of_day='GeoLocal',
                 return_boundaries=False,
                 sum_shape=True,
                 shape_units=None,
                 output_name='summarize nearby' + str(datetime.now().microsecond))
```

```python
item
```

```python
assert isinstance(item, Item)
```

```python
item.delete()
```

```python
item = summarize_nearby(sum_nearby_layer=item2.layers[0],
                 summary_layer=item1.layers[0],
                 near_type='StraightLine',
                 distances=[5],
                 units='Miles',
                 time_zone_for_time_of_day='GeoLocal',
                 return_boundaries=False,
                 sum_shape=True,
                 shape_units=None,
                 future=True,
                 output_name='summarize nearby' + str(datetime.now().microsecond))
```

```python
item.result()
```

```python
assert isinstance(item, GPJob)
assert isinstance(item.result(), Item)
item.result().delete()
```

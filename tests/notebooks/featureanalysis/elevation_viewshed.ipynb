```python
from arcgis.gis import GIS
gis = GIS('https://geosaurus.maps.arcgis.com', 'arcgis_python', 'P@ssword123')
```

```python
from arcgis.features.elevation import viewshed
```

```python
hq_item = gis.content.get('c7665d3c8e6f48a79f07b79677996bed')
hq_lyr = hq_item.layers[0]
```

```python
hq_item
```

```python
fs = hq_lyr.query()
```

```python
fs
```

```python
test = viewshed(input_points=fs,
         maximum_distance=2,
         maximum_distance_units='Miles',
         observer_height=6,
         observer_height_units='Feet',
         surface_offset=100,
         surface_offset_units='Meters',
         generalize_viewshed_polygons=True,
               future=True)
test
```

```python
test.result()
```

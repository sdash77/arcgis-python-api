## Test against AGO

```python
from datetime import datetime
from arcgis.features import FeatureCollection
from arcgis.gis import Item
from arcgis.geoprocessing._job import GPJob
```

```python
from arcgis.gis import GIS
```

```python
gis = GIS('https://www.arcgis.com', 'arcgis_python', 'amazing_arcgis_123')
```

Delete old output

```python
try:
    for i in gis.content.search("RDVehicleRoutingProblem.zip"):
        print(i.delete())
except Exception as ex:
    print(str(ex))
```

```python
from pathlib import Path
import os

file_name = "RDVehicleRoutingProblem.zip"
fp = Path(os.getcwd(), file_name)
```

Upload local fgdb zip

```python
from datetime import datetime

item = gis.content.add(item_properties={
                            "title" : "rdata" + str(datetime.now().microsecond),
                            'type' : "File Geodatabase",
                            'tags' : "A,B,C"
                       }, 
                       data=str(fp))
item
```

```python
from arcgis.features.manage_data import create_route_layers
```

## Create route layers - synchronous

```python
%%time
test1 = create_route_layers(route_data_item=item,
                            delete_route_data_item=False,
                            tags="test",
                            summary="check",
                            route_name_prefix="pri",
                            folder_name="route_layer_tool")
```

```python
assert isinstance(test1, list)
```

```python
test1
```

```python
[i.delete() for i in test1]
```

## Create route layers - async

```python
%%time
test2 = create_route_layers(route_data_item=item,
                            delete_route_data_item=False,
                            tags="test",
                            summary="check",
                            route_name_prefix="pri",
                            future=True,
                            folder_name="route_layer_tool")
assert isinstance(test2, GPJob)
```

```python
assert isinstance(test2.result(), list)
[i.delete() for i in test2.result()]
```

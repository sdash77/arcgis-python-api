```python
from arcgis.gis import GIS
from arcgis.features import FeatureCollection, FeatureSet
from arcgis.gis import Item
from arcgis.geoprocessing._job import GPJob

gis = GIS("https://geosaurus.maps.arcgis.com", "arcgis_python", "P@ssword123")
```

```python
from arcgis.features.elevation import summarize_elevation
```

```python
item = gis.content.get('27908e5dc391490d823a10c9420ce1f8')
```

```python
lyr = item.layers[0]
```

```python
fs = lyr.query()
```

```python
fs
```

```python
test = summarize_elevation(input_features=fs,
                           dem_resolution='FINEST',
                           include_slope_aspect=True)
```

```python
assert isinstance(test, FeatureSet)
```

```python
test = summarize_elevation(input_features=fs)
```

```python
assert isinstance(test, FeatureSet)
test
```

```python
test = summarize_elevation(input_features=fs, future=True)
test = test.result()
```

```python
assert isinstance(test, FeatureSet)
```

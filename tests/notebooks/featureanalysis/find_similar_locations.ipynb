```python
from datetime import datetime
from arcgis.features import FeatureCollection, FeatureSet
from arcgis.gis import Item
from arcgis.geoprocessing._job import GPJob
```

```python
from arcgis.gis import GIS
#gis = GIS(profile="your_online_profile")
gis = GIS('http://deldev.maps.arcgis.com', 'demos_deldev', 'DelDevs.123')
```

```python
from arcgis.features.find_locations import find_similar_locations
```

```python
item1 = gis.content.get('05df66e50bdb426ca1326d0123fc49e1')
```

```python
item2 = gis.content.get('393fc9147f484ee6a9f7c7352dacbb2d')
```

```python
top_4_most_similar_results = find_similar_locations(item1.layers[0],item2.layers[0],
                                                    analysis_fields=['THH17','THH35','THH02','THH05','POPDENS14','FAMGRW10_14','UNEMPRT_CY'],
                                                    number_of_results=4)
```

```python
assert isinstance(top_4_most_similar_results, dict)
assert isinstance(top_4_most_similar_results['similar_result_layer'], FeatureCollection)
```

```python
from datetime import datetime
top_4_most_similar_results = find_similar_locations(item1.layers[0],item2.layers[0],
                                                    analysis_fields=['THH17','THH35','THH02','THH05','POPDENS14','FAMGRW10_14','UNEMPRT_CY'],
                                                    output_name = "top 4 similar locations" + str(datetime.now().microsecond),
                                                    number_of_results=4)
```

```python
assert isinstance(top_4_most_similar_results, dict)
```

```python
assert isinstance(top_4_most_similar_results['similar_result_layer'], Item)
```

```python
assert top_4_most_similar_results['similar_result_layer'].delete()
```

#### Test Overwrite

```python
#import sys
#sys.path.insert(0, r"C:\\ipython_workfolder\\geosaurus\\src")
from arcgis.gis import GIS

portal = GIS(profile="your_portal_profile", verify_cert=False,)
```

```python
airport_item = portal.content.get("8c3a85d4032049ff8333965c6a2a6219")
airport_layer = airport_item.layers[2]
seaport_layer = airport_item.layers[5]
```

```python
test = find_similar_locations(
    airport_layer,
    seaport_layer,
    analysis_fields=["ELEV_FEET", "PASSENGERS"],
    number_of_results=4,
    output_name="similar_locations_overwrite",
)
```

```python
# test is a dictionary and we want to get only the layer to overwrite
fl_to_overwrite = portal.content.get(test["similar_result_layer"].id).layers[0]
print(fl_to_overwrite)
print(test["similar_result_layer"].id)
```

```python
overwrite = find_similar_locations(
    airport_layer,
    seaport_layer,
    analysis_fields=["ELEV_FEET"],
    number_of_results=5,
    output_name=fl_to_overwrite,
    context={"overwrite": True},
)
```

```python
# Notice the ids are the same, if overwrite fails then fl is appended instead
print(overwrite["similar_result_layer"].layers)
print(overwrite["similar_result_layer"].id)
```

```python
overwrite["similar_result_layer"].delete()
```

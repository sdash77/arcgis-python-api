```python
from datetime import datetime
from arcgis.features import FeatureCollection, FeatureSet
from arcgis.gis import Item
from arcgis.geoprocessing._job import GPJob

from arcgis.gis import GIS
gis = GIS('http://deldev.maps.arcgis.com', 'demos_deldev', 'DelDevs12')
```

```python
from arcgis.features.hydrology import trace_downstream
```

```python
item = gis.content.get('1f163950afbb4bf39f3ee67cf411761d')
```

```python
item
```

```python
fs = item.layers[0].query()
```

```python
test = trace_downstream(input_points=fs,
                 source_database='Finest',
                 generalize=False)
```

```python
assert isinstance(test, FeatureSet)
test
```

```python
test = trace_downstream(input_points=fs,
                 source_database='Finest',
                        future=True,
                 generalize=False)
```

```python
assert isinstance(test, GPJob)
assert isinstance(test.result(), FeatureSet)
```

```python
df = item.layers[0].query(as_df=True)
```

```python
test = trace_downstream(input_points=df,
                 source_database='Finest',
                 generalize=False)
```

```python
import pandas as pd
assert type(test.sdf) == pd.DataFrame
```

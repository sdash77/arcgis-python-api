```python
from datetime import datetime

from arcgis.gis import GIS
#gis = GIS(profile="your_online_profile")
```

```python
gis = GIS("https://deldev.maps.arcgis.com", "demos_deldev", "DelDevs12")
```

```python
items = gis.content.search('tags:cougar_habitat', 'Feature Layer')
```

```python
items
```

```python
state_park = items[0].layers[0]
vegetation = items[1].layers[0]
slope = items[2].layers[0]
sub_watershed = items[3].layers[0]
highways = items[4].layers[0]
streams = items[5].layers[0]
```

```python
from arcgis.features.find_locations import derive_new_locations
```

```python
new_loaction = derive_new_locations(input_layers=[slope, vegetation, streams, highways],
                                    expressions=[{"operator":"","layer":0,"selectingLayer":1,"spatialRel":"intersects"},
                                                 {"operator":"and","layer":0,"selectingLayer":2,"spatialRel":"withinDistance","distance":500,"units":"Feet"},
                                                 {"operator":"and","layer":0,"selectingLayer":3,"spatialRel":"notWithinDistance","distance":1500,"units":"Feet"},
                                                 {"operator":"and","layer":0,"where":"GRIDCODE = 1"}], future=True)
                                    #output_name='derive_new_loactionsvb' + str(datetime.now().microsecond))
```

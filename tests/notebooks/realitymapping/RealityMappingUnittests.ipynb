### Create Reality Mapping Project

```python
def currentTime():
    return time.strftime("%Y%m%d%H%M%S", time.localtime())
```

```python
rm_project = realitymapping.Project(project="rmProject" + currentTime(), gis=gis)
```

### Create Mission

```python
imageList = glob.glob(os.path.join(imageFolderPath, '*.JPG'), recursive=True)
imageList
```

```python
mission_name = "mission" + currentTime()
image_collection_name = "imgcollect" + currentTime()

mission_1 = rm_project.create_mission(image_list=imageList,
  mission_name=mission_name,
  image_collection_name=image_collection_name,
  raster_type_name="UAV/UAS",
  raster_type_params={"isAltitudeFlightHeight":"false",
  "averagezdem": {"url": "https://elevation3d.arcgis.com/arcgis/rest/services/WorldElevation3D/Terrain3D/ImageServer"}}
)
 
```

### Alter Processing States


```python
job_aps = mission_1.alter_processing_states(new_states={"blockadjustment": "raw","dem": "Dense_Natual_Neighbor","seamlines":"VORONOI","colorcorrection":"SingleColor"}, 
                                            gis=gis, future=True)
assert job_aps
assert job_aps.result()
```

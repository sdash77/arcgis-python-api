```python
# Necessary imports
from arcgis.apps.dashboard import Details, Dashboard

from arcgis.gis import GIS
gis = GIS('https://geosaurus.maps.arcgis.com', 'arcgis_python', 'amazing_arcgis_123')
```

```python
#get the desired item
item =  gis.content.get('2eb03d3d8106443ba54a6b8a7d1b0d11')
```

```python
#display the item
item
```

```python
#display the item type
item.layers
```

```python
#create the 'Details' element object 
element = Details(item=item, max_features_displayed=5, layer=4)
```

```python
element.description = 'testme'
```

```python
#visualize the element to see if you want to make any changes to it
element
```

```python
element.width = 2 #value to be from 0 to 1
```

```python
# Details element is used to display a layer's pop-up information. Each pop-up has four distinct parts: 
#title, content, media, and attachments. When configuring the details element, each of these parts can be switched on or off.
#By default all four are set to True.
element.show_title = True
element.show_content = True
element.show_media = True
element.show_attachment = True
```

```python
element.show_last_update = False
```

```python
element
```

```python
# check the default number of features to display on map
element.max_features
```

```python
#you could change this default based on your choice. Since it's a setter, you cna directly set the value.
#i'm setting it same as the number of rows in the layer
element.max_features = len(item.layers[0].query(as_df=True))
```

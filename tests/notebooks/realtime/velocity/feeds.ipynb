```python
sample_feed = None
try:
    sample_feed = feeds.get("e4d3c42193b14b48b912306919617010")
    print(sample_feed)
except Exception as e:
    print("Error while fetching the feed: " + e.__str__())
```

- Property: `<item>.status`
- Retrieve status of `sample_feed`

```python
try:
    status = sample_feed.status
    print(status)
except Exception as e:
    print("Error while fetching the feed status: " + e.__str__())
```

```python
try:
    print(sample_feed.start())
except Exception as e:
    print("Error while starting the feed: " + e.__str__())
```

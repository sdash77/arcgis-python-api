- Get instance of feeds_manager from `velocity`

```python
velocity = gis.velocity
velocity
```

```python
feeds_manager = velocity.feeds_manager
feeds_manager
```

- Get list feeds configurations

```python
all_feeds = feeds_manager.items
all_feeds
```

```python
sample_feed.start()
```

## Configure the Geotab Feed
Configuring a delimited format feed that contains location, time, track id and info about satellite locations.

```python
from arcgis.realtime.velocity.feeds import Geotab
from arcgis.realtime.velocity.feeds.geometry import XYZGeometry, SingleFieldGeometry
from arcgis.realtime.velocity.feeds.time import TimeInterval, TimeInstant

# Geotab Properties
name = "geotab_feed_1"
description = "some description about the Geotab feed"

geotab = Geotab(
    label=name,
    description=description,
    url="https://my.geotab.com/apiv1",
    database="some_database_name_sd",
    username="some_user_name",
    password="some_password",
    data_format=None
)



# # all properties can also be defined right away in the constructor as follows
#
# time = TimeInstant(
#     interval_start_field="dateTime"
# )
# geometry = XYZGeometry("longitude", "latitude", 4326)
#
# geotab = Geotab(
#     label=name,
#     description=description,
#     url="https://my.geotab.com/apiv1",
#     database="some_database_name_sd",
#     username="some_user_name",
#     password="some_password",
#     data_format=None,
#     geometry=geometry,
#     time=time,
# )
```

### Manipulate the schema - rename or remove fields, change field data-type
Renaming `vehicleName` to `vehicleName1`

```python
geotab.rename_field("vehicleName", "vehicleName1")
```

### Set track id field
Set `device_id` (Maritime Mobile Service Identity) as the track-id

```python
geotab.set_track_id("device_id")
```

Set XY geometry

```python
xy_geometry = XYZGeometry("longitude", "latitude", 4326)
geotab.set_geometry_config(xy_geometry)
```

Set Time field

```python
geotab.set_time_config(TimeInstant("dateTime"))
```

### Create the Feed (and start it).

```python
geotab_feed = feeds.create(geotab)
#geotab_feed.start()
feeds.items
```

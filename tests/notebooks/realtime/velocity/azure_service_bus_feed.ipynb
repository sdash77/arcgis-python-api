### Manipulate the schema - rename or remove fields, change field data-type
Renaming fields to timestamp,route_id,latitude,longitude,route_pt_sequence

```python
azure_service_bus_config.rename_field("field1", "timestamp")
azure_service_bus_config.rename_field("field2", "route_id")
azure_service_bus_config.rename_field("field3", "latitude")
azure_service_bus_config.rename_field("field4", "longitude")
azure_service_bus_config.rename_field("field5", "route_pt_sequence")
```

### Set track id field
Set `field2` (route_id) as the track-id

```python
azure_service_bus_config.set_track_id("field2")
```

Set XY geometry

```python
xy_geometry = XYZGeometry("field4", "field3", 4326)
azure_service_bus_config.set_geometry_config(xy_geometry)
```

Set `timestamp` as the time field

```python
time_config = TimeInstant("field1")
azure_service_bus_config.set_time_config(time_config)
```

### Create the Feed (and start it).

```python
azure_service_bus_feed = feeds.create(azure_service_bus_config)
#azure_service_bus_feed.start()
feeds.items
```

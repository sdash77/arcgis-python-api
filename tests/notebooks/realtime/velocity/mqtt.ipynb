### Manipulate the schema - rename or remove fields, change field data-type
Renaming `VesselName` to `VesselName1`

```python
mqtt_config.rename_field("VesselName", "VesselName1")
```

### Set track id field
Set `MMSI` (Maritime Mobile Service Identity) as the track-id

```python
mqtt_config.set_track_id("MMSI")
```

Set XY geometry

```python
xy_geometry = XYZGeometry("LON", "LAT", 4326)
mqtt_config.set_geometry_config(xy_geometry)
```

### Create the Feed (and start it).

```python
mqtt_feed = feeds.create(mqtt_config)
#mqtt_feed.start()
feeds.items
```

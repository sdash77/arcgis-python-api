- Get instance of bigdata_analytics from `velocity`

```python
velocity = gis.velocity
velocity
```

```python
bigdata_analytics = velocity.bigdata_analytics
bigdata_analytics
```

- Get list of BigData Analytics configurations

```python
all_bigdata_analytics = bigdata_analytics.items
all_bigdata_analytics
```

- Method: `<item>.get(id)`
- Get BigData Analytics by id

```python
sample_bigdata_task = bigdata_analytics.get("bf9c78e94fde40dabda67b6b4c8d3395")
sample_bigdata_task
```

- Property: `<item>.status`

```python
sample_bigdata_task.status
```

```python
sample_bigdata_task.start()
```

- Method: `<item>.stop()`

```python
sample_bigdata_task.stop()
```

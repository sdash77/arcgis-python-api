- Property: `<item>.status`

```python
try:
    print(sample_bigdata_task.status)
except Exception as e:
    print("Error while fetching the task status: " + e.__str__())
```

```python
try:
    print(sample_bigdata_task.start())
except Exception as e:
    print("Error while starting the task: " + e.__str__())
```

```python
try:
    print(sample_bigdata_task.stop())
except Exception as e:
    print("Error while stopping the task: " + e.__str__())
```

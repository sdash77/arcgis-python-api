```python
from arcgis.gis import GIS
```

```python
# Enter Url and Credentials
gis = GIS(url="", username="", password="")
```

- Get instance of realtime_analytics_manager from `velocity`

```python
velocity = gis.velocity
velocity
```

```python
realtime_analytics_manager = velocity.realtime_analytics_manager
realtime_analytics_manager
```

- Get list of Realtime Analytics configurations

```python
all_realtime_analytics = realtime_analytics_manager.items
all_realtime_analytics
```

- Method: `<item>.get(id)`
- Get RealTime Analytics by id

```python
sample_realtime_task = realtime_analytics_manager.get("7a8f2100aedb47be85d57d4969757a16")
sample_realtime_task
```

- Property: `<item>.status`
- Retrieve status of `sample_realtime_task`

```python
sample_realtime_task.status
```

- Method: `<item>.start()`
- Start Realtime Analytics

```python
sample_realtime_task.start()
```

- Method: `<item>.stop()`
- Stop Realtime Analytics

```python
sample_realtime_task.stop()
```

- Property: `<item>.metrics`
- Retrieve metrics of `sample_realtime_task`

```python
sample_realtime_task.metrics
```

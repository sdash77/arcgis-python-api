## Configure the Stream Layer Feed

Feed Portal ID: 19ae6ce683944a12b4034106dcf42164

```python
from arcgis.realtime.velocity.feeds import StreamLayer
from arcgis.realtime.velocity.feeds.time import TimeInterval, TimeInstant


# Stream Layer Properties
name = "stream_layer_config_feed_1"
description = "some description about the Stream Layer feed"
portal_id = "19ae6ce683944a12b4034106dcf42164"
extent = {
    "spatialReference": {
        "latestWkid": 3857,
        "wkid": 102100
    },
    "xmin": -14784278.027601289,
    "ymin": 2604610.848073723,
    "xmax": -11451317.846255329,
    "ymax": 6852675.132049575
}
stream_layer_config = StreamLayer(
    label=name,
    description=description,
    portal_item_id=portal_id,
    query="1=1",
    fields="*",
    outSR=4326,
    extent=extent
)

stream_layer_config


# # all properties can also be defined right away in the constructor as follows
#
# time = TimeInterval(
#     interval_start_field="pubDate",
#     interval_end_field="updated"
# )
#
# run_interval = RunInterval(
#     cron_expression="0 * * ? * * *",
#     timezone="America/Los_Angeles"
# )
#
# stream_layer_config1 = HttpPoller(
#     label=name,
#     description=description,
#     url=url,
#     http_method="GET",
#     http_auth_type=http_auth,
#     track_id_field="link",
#     data_format=data_format,
#     geometry=geometry,
#     time=time,
#     run_interval=run_interval
# )
```

### Manipulate the schema - rename or remove fields, change field data-type
1. Renaming `EstArrivalTime` to `ETA`

```python
stream_layer_config.rename_field("EstArrivalTime", "ETA")
```

### Set track id field
Set `FlightID` as the track-id

```python
stream_layer_config.set_track_id("FlightID")
```

### Set time field

```python
# time interval
#time = TimeInterval(interval_start_field="DateTimeStamp", interval_end_field="ArrivalTime")

# time instant
time = TimeInstant(time_field="DateTimeStamp")


stream_layer_config.set_time_config(time=time)
```

### Create the Feed (and start it).

```python
stream_layer_feed = feeds.create(stream_layer_config)
#stream_layer_feed.start()
feeds.items
```

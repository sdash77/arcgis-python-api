## Configure the MQTT Feed
Configuring a delimited format feed that contains location, time, track id and info about satellite locations.

```python
from arcgis.realtime.velocity.feeds import Kafka
from arcgis.realtime.velocity.feeds.kafka_authentication_type import NoAuth, SASLPlain
from arcgis.realtime.velocity.feeds.geometry import XYZGeometry, SingleFieldGeometry
from arcgis.realtime.velocity.feeds.time import TimeInterval, TimeInstant

# Kafka Properties
name = "kafka_feed_1"
description = "some description about the Kafka feed"

kafka_config = Kafka(
    label=name,
    description=description,
    brokers="kafka.a4iot.com:9092",
    topics="ais_json_1",
    authentication=NoAuth(),
    data_format=None
)



# # all properties can also be defined right away in the constructor as follows
#data_format = DelimitedFormat("\n", ",", False)
# geometry = XYZGeometry(
#     x_field="LON",
#     y_field="LAT",
#     wkid=4326,
#     z_field="ALT",
#     z_unit="Meters"
# )
#
# time = TimeInterval(
#     interval_start_field="pubDate",
#     interval_end_field="updated"
# )
#
#
# kafka_config = Kafka(
#     label=name,
#     description=description,
#     brokers="kafka.a4iot.com:9092",
#     topics="ais_json_1",
#     data_format=data_format,
#     geometry=geometry,
#     time=time,
# )
```

### Manipulate the schema - rename or remove fields, change field data-type
Renaming `VesselName` to `VesselName1`

```python
kafka_config.rename_field("VesselName", "VesselName1")
```

### Set track id field
Set `MMSI` (Maritime Mobile Service Identity) as the track-id

```python
kafka_config.set_track_id("MMSI")
```

Set XY geometry

```python
xy_geometry = XYZGeometry("LON", "LAT", 4326)
kafka_config.set_geometry_config(xy_geometry)
```

### Create the Feed (and start it).

```python
kafka_feed = feeds.create(kafka_config)
#kafka_feed.start()
feeds.items
```

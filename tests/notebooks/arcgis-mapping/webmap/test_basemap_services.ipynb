```python
from arcgis.gis import GIS
from arcgis.map import Map
from arcgis.layers import BasemapService
import time

gis = GIS(profile="your_online_profile")
webmap = Map()

webmap
```

## Test Styles

```python
basemap_service = BasemapService()
all_services = basemap_service.styles()
len(all_services)
```

```python
for name, path in all_services.items():
    webmap.basemap.basemap = {"id": path}
    time.sleep(5)
print("Done")
```

```python
# This will be the last in the list:
print("Last basemap: " + path)
print("  ")
webmap.basemap.basemap
```

## Test with language

webmap1 = Map()

webmap1

```python
basemap_service = BasemapService()
all_languages = basemap_service.languages
len(all_languages)
```

```python
for lang_dict in all_languages:
    webmap1.basemap.basemap = {"id": "arcgis/streets", "language": lang_dict["code"]}
    time.sleep(5)
print("Done")
```

```python
# notice the customParams key in the basemap layers
webmap1.basemap.basemap
```

## Test with Places

```python
webmap2 = Map()
webmap2
```

```python
basemap_service = BasemapService()
place_options = basemap_service.places
len(place_options)
```

```python
place_options
```

```python
for option in place_options:
    webmap2.basemap.basemap = {"id": "arcgis/streets", "places": option["code"]}
    time.sleep(5)
print("Done")
```

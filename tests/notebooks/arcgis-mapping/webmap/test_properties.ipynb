# Test WebMap Properties

#### The purpose of this lab is to test the different properties on a widget. In order to test correctly, you need to visually ensure that the properties are being added/changed to the widget, not just run the notebook all at once.

### Import libraries

```python
from arcgis.map import Map
from arcgis.gis import GIS

gis = GIS(profile="your_online_profile")
```

### Create Empty WebMap

```python
webmap = Map()
webmap
```

```python
# Zoom: Map should change zoom here
zoom = webmap.zoom
webmap.zoom = 10
assert webmap.zoom != zoom
```

```python
# Extent: Map should change location here
extent = webmap.extent
webmap.extent = {
    "spatialReference": {"wkid": 102100},
    "xmin": -14024242.199180739,
    "ymin": 5803295.297047005,
    "xmax": -13065416.116371742,
    "ymax": 6292492.278072002,
}
assert webmap.extent != extent
```

```python
# Layers: empty list right now
assert webmap.content.layers == []
```

```python
# Turn on widgets
webmap.layer_list.enabled = True
webmap.legend.enabled = True
webmap.bookmarks.enabled = True

# Should see all three widgets in top right corner!
```

```python
# Basemaps
webmap.basemap.basemaps
```

```python
# Basemaps Gallery
webmap.basemap.basemap_gallery
```

```python
# Basemap
import time

for basemap in webmap.basemap.basemaps:
    webmap.basemap.basemap = basemap
    time.sleep(3)
```

# Test Add Layers to WebMap

#### The purpose of this lab is to test the ability to add different types of layers to a widget. In order to test correctly, you need to visually ensure that the layers are being added to the widget, not just run the notebook all at once.

### Import Libraries

```python
from arcgis.map.map_widget import Map
from arcgis.map import GroupLayer
from arcgis.features import FeatureLayer
from arcgis.layers import VectorTileLayer, MapServiceLayer, CSVLayer, KMLLayer, WMSLayer, GeoJSONLayer
from arcgis.features import FeatureCollection
from arcgis.raster import ImageryLayer
from arcgis.gis import GIS
```

```python
gis = GIS('https://sha-arcgis-ra.esri.com/portal', 'admin', 'adminadmin', verify_cert=False)
```

### Create WebMap

```python
webmap = Map('Amberg')
webmap.basemap.basemap = "topo-vector"
webmap
```

```python
nl.properties
```

```python
webmap.legend.enabled = True
webmap.layer_list.enabled = True
```

```python
# Feature Layer
feature_layer = FeatureLayer(
    "https://sampleserver6.arcgisonline.com/arcgis/rest/services/Census/MapServer/3"
)
webmap.content.add(feature_layer)
```

```python
# Vector Tile Layer
vector_layer = VectorTileLayer(
    "https://basemaps.arcgis.com/arcgis/rest/services/World_Basemap_v2/VectorTileServer"
)
webmap.content.add(vector_layer)
```

```python
webmap.layer_list.enabled = True
```

```python
# CSV Layer
csv_layer = CSVLayer(
    "https://earthquake.usgs.gov/earthquakes/feed/v1.0/summary/2.5_week.csv"
)
webmap.content.add(csv_layer)
```

```python
# KML Layer
kml_layer = KMLLayer("http://quickmap.dot.ca.gov/data/lcs.kml")
webmap.content.add(kml_layer)
```

```python
# GeoJSON Layer
geojson_layer = GeoJSONLayer(
    "https://earthquake.usgs.gov/earthquakes/feed/v1.0/summary/all_month.geojson"
)
webmap.content.add(geojson_layer)
```

```python
# Feature Set
layer = FeatureLayer(
    "https://sampleserver6.arcgisonline.com/arcgis/rest/services/Census/MapServer/3"
)
feature_set = layer.query()
webmap.content.add(feature_set)
```

```python
# Feature Collection
layer = FeatureLayer(
    "https://sampleserver6.arcgisonline.com/arcgis/rest/services/Census/MapServer/3"
)
fs = layer.query("OBJECTID < 20")
fc = FeatureCollection.from_featureset(fs)
webmap.content.add(fc)
```

```python
# SEDF
layer = FeatureLayer(
    "https://sampleserver6.arcgisonline.com/arcgis/rest/services/Census/MapServer/3"
)
sedf = layer.query(as_df=True)
webmap.content.add(sedf)
```

```python
# Group Layer
item = gis.content.get("b84064f8638c47e89bfd3edb49acb628")
webmap.content.add(item)
```

```python
# Imagery Layer
layer = ImageryLayer(
    "https://sha-arcgis-ra.esri.com/server/rest/services/090160_tif/ImageServer", gis=gis
)
webmap.content.add(layer)
```

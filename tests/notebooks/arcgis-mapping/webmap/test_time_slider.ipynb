## Test Time Slider

```python
from arcgis.gis import GIS
from arcgis.map import Map

gis = GIS(verify_cert=False)
```

```python
g_map = Map(location="Mexico")
g_map
```

```python
item = gis.content.get("f14573e9c5ce4b41aecb199d5723209b")
g_map.content.add(item)
```

```python
g_map.time_slider.enabled = True
```

```python
import datetime as dt

g_map.time_slider.set_time_extent(dt.datetime(2000, 1, 2), dt.datetime(2002, 1, 5))
```

## Relationship Renderer

```python
from arcgis.gis import GIS
from arcgis.map import Map

gis = GIS(profile="your_online_profile")
```

```python
item = gis.content.get("c514eddc6d584e85bc2f90be25305fc8")
```

```python
mmap = Map()
mmap
```

```python
mmap.content.add(item.layers[1])  # add state
smm = mmap.content.renderer(0).smart_mapping
```

```python
smm.relationship_renderer({"field": "v001_rawvalue"}, {"field": "v002_rawvalue"})
```

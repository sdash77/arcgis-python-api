## Univariate Color Size Renderer 

```python
from arcgis.gis import GIS
from arcgis.map import Map

gis = GIS()
```

```python
item = gis.content.get("14c5450526a8430298b2fa74da12c2f4")
for layer in item.layers:
    print(layer.properties.name)
```

```python
mmap = Map(location="California")
mmap
```

```python
mmap.content.add(item.layers[0])
```

```python
smm = mmap.content.renderer(0).smart_mapping
```

```python
smm.univariate_color_size_renderer(
    field="POPULATION",
    normalization_field="POP_SQMI",
)
```

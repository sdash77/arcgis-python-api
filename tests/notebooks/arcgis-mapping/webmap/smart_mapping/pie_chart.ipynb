# Pie Chart Renderer

```python
from arcgis.gis import GIS
from arcgis.map.map_widget import Map

gis = GIS(profile="your_online_profile")
```

```python
item = gis.content.get("eb03a1db09434a658e283c9fd55cdfc5")
```

```python
mmap = Map(location="Rome")
mmap
```

```python
mmap.content.add(item.layers[1])  # add state
smm = mmap.content.renderer(0).smart_mapping
```

```python
smm.pie_chart_renderer([{"field": "route_id"}])
```

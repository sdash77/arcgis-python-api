# Predominance Renderer

```python
from arcgis.gis import GIS
from arcgis.map.map_widget import Map
from arcgis.features import FeatureLayer

gis = GIS(profile="your_online_profile")
```

```python
mmap = Map()
mmap
```

```python
fl = FeatureLayer(
    "https://services.arcgis.com/V6ZHFr6zdgNZuVG0/arcgis/rest/services/USA_County_Crops_2007/FeatureServer/0"
)
fl
```

```python
mmap.content.add(fl)
smm = mmap.content.renderer(0).smart_mapping
```

```python
smm.predominance_renderer(
    [
        {"name": "M217_07", "label": "Vegetables"},
        {"name": "M188_07", "label": "Cotton"},
        {"name": "M172_07", "label": "Wheat"},
        {"name": "M193_07", "label": "Soybeans"},
        {"name": "M163_07", "label": "Corn"},
    ],
    include_opacity_variable=True,
)
```

# Dot Density Renderer

```python
from arcgis.gis import GIS
from arcgis.map.map_widget import Map

gis = GIS(profile="your_online_profile")
```

```python
item = gis.content.get("14c5450526a8430298b2fa74da12c2f4")
```

```python
mmap = Map()
mmap
```

```python
mmap.content.add(item.layers[0])  # add state
smm = mmap.content.renderer(0).smart_mapping
```

```python
smm.dot_density_renderer([{"field": "POPULATION"}])
```

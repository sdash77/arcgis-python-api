```python
from arcgis.map.map_widget import Map
from arcgis.raster import ImageryLayer, Raster
from arcgis.gis import GIS
```

```python
gis = GIS('https://sha-arcgis-ra.esri.com/portal', '', '', verify_cert=False)
```

#Testing add RFT output layer to map widget

```python
webmap = Map('Amberg')
webmap.basemap.basemap = "topo-vector"
webmap
```

```python
# Imagery Layer
layer = ImageryLayer(
    "https://sha-arcgis-ra.esri.com/server/rest/services/090160_tif/ImageServer", gis=gis
)
webmap.content.add(layer)
```

```python
from arcgis.raster.functions import RFT
rft_item =  gis.content.get('870941f10ff843d38d54e53d514c3974')
rft_item
```

```python
rft_obj = RFT(gis.content.get('870941f10ff843d38d54e53d514c3974'))
ras = rft_obj(Raster=layer)
ras
```

```python
webmap.content.add(ras)
```

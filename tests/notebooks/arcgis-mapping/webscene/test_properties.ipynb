# Test WebMap Properties

#### The purpose of this lab is to test the different properties on a widget. In order to test correctly, you need to visually ensure that the properties are being added/changed to the widget, not just run the notebook all at once.

### Import libraries

```python
from arcgis.map import Scene
from arcgis.gis import GIS

gis = GIS(profile="your_online_profile")
```

### Create Empty WebMap

```python
webscene = Scene()
webscene
```

```python
# Zoom: Map should change zoom here
zoom = webscene.zoom
webscene.zoom = 10
assert webscene.zoom != zoom
```

```python
# Extent: Map should change location here
extent = webscene.extent
webscene.extent = {
    "spatialReference": {"wkid": 102100},
    "xmin": -14024242.199180739,
    "ymin": 5803295.297047005,
    "xmax": -13065416.116371742,
    "ymax": 6292492.278072002,
}
assert webscene.extent != extent
```

```python
# Layers: empty list right now
assert webscene.layers == []
```

```python
# Turn on widgets
webscene.layer_list.enabled = True
webscene.legend.enabled = True

# Should see both widgets in top right corner!
```

```python
# Basemaps
webscene.basemaps
```

```python
# Basemaps Gallery
webscene.basemap_gallery
```

```python
# Basemap
import time

for basemap in webscene.basemaps:
    webscene.basemap = basemap
    time.sleep(5)
```

```python
from arcgis.gis import GIS
from arcgis.map import Scene

gis = GIS(profile="your_online_profile")
```

```python
scene = Scene()
scene
```

```python
scene.zoom = 5
```

```python
# Check the 3d basemaps available
bm_3d = scene.basemap.basemaps3d
bm_3d
```

```python
# Current basemap
scene.basemap.basemap
```

```python
# Change to each 3d Basemap
from time import sleep


for bm in bm_3d:
    scene.basemap.basemap = bm
    sleep(5)

# This should be last basemap in list now
scene.basemap.basemap
```

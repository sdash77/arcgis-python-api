```python
from arcgis.gis import GIS, Group, GroupManager
import pandas as pd
import pytest
```

```python
profiles = ['your_online_profile', 'your_enterprise_profile']
gis_objects = {}
for profile in profiles:
    gis = GIS(profile=profile, verify_cert=False)
    gis_objects[profile] = gis
```

### Group Manager Methods

```python
from datetime import datetime
for profile in profiles:
    gis = gis_objects[profile]
    assert isinstance(gis.users.me.groups, list)
    gm = gis.groups
    assert isinstance(gm, GroupManager)
    assert isinstance(gm.search("*"), list)
    if len(gis.users.me.groups) > 0:
        assert isinstance(gm.get(gis.users.me.groups[0].id), Group)
    else:
        assert isinstance(gm.get(gm.search("*")[0].id), Group)
    title = "eraseme" + str(datetime.now().microsecond)[:3]
    tags = 'erase me, test'
    grp = gm.create(title=title, tags=tags)
    assert isinstance(grp, Group)
    assert grp.delete()
```

### Group Methods

```python
for profile in profiles:
    gis = gis_objects[profile]
    gm = gis.groups
    title = "eraseme" + str(datetime.now().microsecond)[:3]
    tags = 'erase me, test'
    grp = gm.create(title=title, tags=tags)
    assert isinstance(grp, Group)
    assert grp.update(tags="philly, nyc, boston")
    assert grp.tags == ['philly', 'nyc', 'boston']
    username = "eraseme" + str(datetime.now().microsecond)[:3]
    um = gis.users
    user = um.create(username=username,
                 password='Strong.B4D',
                 firstname='delete',
                 lastname='me',
                 email='fakeesri@esri.com')
    assert grp.add_users([user])
    assert len(grp.get_members()['users']) == 1
    assert grp.remove_users([user])
    assert len(grp.get_members()['users']) == 0
    assert user.delete()
    assert grp.delete()
```

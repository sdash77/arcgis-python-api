### Spot Checks the ArcGIS Server for Common Access Tools

```python
from arcgis.gis import GIS
import pandas as pd
import pytest
```

```python
profiles = ['your_enterprise_profile']
gis_objects = {}
for profile in profiles:
    gis = GIS(profile=profile, verify_cert=False)
    gis_objects[profile] = gis
```

### Access the Servers on Portal

```python
from datetime import datetime
from arcgis.gis.server import ServerManager, Server
```

```python
for profile in profiles:
    gis = gis_objects[profile]
    sm = gis.admin.servers
    assert isinstance(sm, ServerManager)
    servers = sm.list()
    assert isinstance(servers, list)
    assert sm.validate() in (True, False)
    assert isinstance(sm.get("HOSTING_SERVER"), list)
```

### Work on a Single Server

```python
from arcgis.gis.server import DataStoreManager, LogManager, MachineManager, SiteManager, SystemManager
```

```python
if len(sm.get("HOSTING_SERVER")) > 0:
    server = sm.get("HOSTING_SERVER")[0]
    assert isinstance(server, Server)
    sd = server.content.list()
    logs = server.logs
    ds = server.datastores
    machines = server.machines
    props = server.properties
    site = server.site
    system = server.system
    assert isinstance(sd, list)
    assert isinstance(machines, MachineManager)
    assert isinstance(logs, LogManager)
    assert isinstance(logs.query(), dict)
    assert isinstance(site, SiteManager)
    assert isinstance(system, SystemManager)
    assert props
    assert isinstance(server.services.list(), list)
    services = server.services.list()
    if len(services) > 0:
        service = services[0]
        assert service.status
        assert service.statistics
    assert isinstance(ds, DataStoreManager)
    assert len(ds.list()) > 0
    assert system.platform_services.properties
```

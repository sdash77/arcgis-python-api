```python
from arcgis.gis import GIS, User, UserManager, RoleManager, Role
import pandas as pd
import pytest
```

```python
profiles = ['your_online_profile', 'your_enterprise_profile']
```

### Basic Login

```python
gis_objects = {}
for profile in profiles:
    gis = GIS(profile=profile, verify_cert=False)
    gis_objects[profile] = gis
    assert isinstance(gis.users.me, User)
```

### UserManager Methods

```python
for profile in profiles:
    gis = gis_objects[profile]
    um = gis.users
    assert isinstance(um, UserManager)
    assert isinstance(um.counts('user_type'), pd.DataFrame)
    assert isinstance(um.get(gis.users.me.username), User)
    assert isinstance(um.license_types, list)
    assert isinstance(um.roles, RoleManager)
    assert isinstance(um.user_groups([gis.users.me]), list)
    assert isinstance(um.search("esri"), list )
```

### Tests the User Management Part of the API

```python
from datetime import datetime
for profile in profiles:
    gis = gis_objects[profile]
    if gis.users.me.role == "org_admin":
        username = "eraseme" + str(datetime.now().microsecond)[:3]
    
        um = gis.users
        if gis._portal.is_arcgisonline:
            user = um.create(username=username,
                 password='Strong.B4D',
                 firstname='delete',
                 lastname='me',
                 email='fakeesri@esri.com',
                 user_type='AdvancedUT')
        else:
            user = um.create(username=username,
                 password='Strong.B4D',
                 firstname='delete',
                 lastname='me',
                 email='fakeesri@esri.com')
        assert isinstance(user, User)
        assert um.disable_users([user])
        assert um.enable_users([user])
        assert user.delete()
```

### Test the User Operations

```python
from datetime import datetime
for profile in profiles:
    gis = gis_objects[profile]
    if gis.users.me.role == "org_admin":
        username = "eraseme" + str(datetime.now().microsecond)[:3]
        um = gis.users
        if gis._portal.is_arcgisonline:
            user = um.create(username=username,
                 password='Strong.B4D',
                 firstname='delete',
                 lastname='me',
                 email='fakeesri@esri.com',
                 user_type='AdvancedUT')
        else:
            user = um.create(username=username,
                 password='Strong.B4D',
                 firstname='delete',
                 lastname='me',
                 email='fakeesri@esri.com')
        assert isinstance(user, User)
        assert user.disable()
        assert user.disabled == True
        assert user.enable()
        assert user.disabled == False
        assert isinstance(gis.users.me.items(), list)
        if gis._portal.is_arcgisonline:
            assert isinstance(gis.users.me.linked_accounts, list)
        assert len(user.items()) == 0
        assert user.delete()
    
```

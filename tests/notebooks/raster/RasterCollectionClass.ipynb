### map

```python
def NDVI(item):
    raster = item['Raster']
    red, nir = raster.get_raster_bands([3, 4])
    ndvi = (nir-red)/(nir+red)
    return {"raster": ndvi, "StdTime": item['AcquisitionDate']}
```

```python
point = geometry.Point({"x":54.9243963,"y":23.9279934,"spatialReference":{"wkid":4326,"latestWkid":4326}})
rc2 = rc.filter_by(where_clause = "SensorName LIKE '%Landsat%'", query_geometry_or_extent = point)
rc3 = rc2.map(NDVI)
rc3
```

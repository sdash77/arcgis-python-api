```python
from arcgis.mapping import WebMap
from arcgis.raster import ImageryLayer

search_result = gis.content.search("World Imagery", max_items=2)
if len(search_result) == 0:
    print("Customized basemap not found!")
else:
    item1 = search_result[0]
    wm_item1 = WebMap(item1)
    for bml in wm_item1.basemap.baseMapLayers:
        print(bml.layerType, bml.url)
        img_lyr = ImageryLayer(bml.url)
        map1.add_layer(img_lyr)
```

```python
from arcgis.features import FeatureLayer

hurricane_layer = FeatureLayer(
    url="https://sampleserver6.arcgisonline.com/arcgis/rest/services/Hurricanes/MapServer/0")
hurricane_map = gis.map('Atlantic Ocean', 3)
hurricane_map.add_layer(hurricane_layer)
hurricane_map
```

```python
hurricane_map.basemap
```

```python
hurricane_map_3D = gis.map('Atlantic Ocean', 3)
hurricane_map_3D.mode = "3D"

renderer = {"renderer": "autocast", #This tells python to use JS autocasting
  "type": "simple",  # JS type of new SimpleRenderer()
  "symbol": {
    "type": "point-3d",  #JS type of new PointSymbol3D()
    "symbolLayers": [{  
      "type": "object",  #JS type of new ObjectSymbol3DLayer()
      "resource": { "primitive": "sphere" },
      "width": 50000
    }]
  },
}

hurricane_map_3D.add_layer(hurricane_layer, renderer)
hurricane_map_3D
```

```python
visual_var_map = gis.map('Atlantic Ocean', 3)
visual_var_map.mode = "3D"

# Update the renderer to include visual variable information
renderer["visualVariables"] = [{ #JS configuration of visualVariables
    "type": "color",
    "field": "PRESSURE", 
    "stops": [
      { "value": 950, "color": "red" },
      { "value": 1020, "color": "blue" }
    ]
  }, {
    "type": "size",
    "field": "WINDSPEED",
    "stops": [
      { "value": 20, "size": 60000 },
      { "value": 150, "size": 500000 }
    ],
    "axis": "height"
  }, {
    "type": "size",
    "axis": "width-and-depth",
    "useSymbolValue": True
  }]
visual_var_map.add_layer(hurricane_layer, renderer)
# visual_var_map.add_layer(hurricane_layer)
visual_var_map
```

```python
visual_var_map.basemap
```

```python
visual_var_map.basemap = 'streets-night-vector'
```

```python
visual_var_map.basemap
```

```python
webscene_properties = {'title':'WS Toprical Cyclones - Summer',
                    'snippet': 'Jupyter notebook widget saved as a web scene',
                    'tags':['automation', 'python']}

webscene_item = visual_var_map.save(webscene_properties)
webscene_item
```

```python
from arcgis.mapping import WebScene
webscene_obj = WebScene(webscene_item)
webscene_obj['baseMap']['baseMapLayers']
```

```python
if 'url' in webscene_obj['baseMap']['baseMapLayers'][0]:
    if "servicesdev.arcgisonline" not in webscene_obj['baseMap']['baseMapLayers'][0]['url']:
        print("[X] New WS's Basemap not loading from devext!")
elif 'styleUrl' in webscene_obj['baseMap']['baseMapLayers'][0]:
    if "servicesdev.arcgisonline" not in webscene_obj['baseMap']['baseMapLayers'][0]['styleUrl']:
        print("[X] New WS's Basemap (style) not loading from devext!")
else:
    pass
```

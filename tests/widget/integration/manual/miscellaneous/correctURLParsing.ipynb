```python
from arcgis.gis import GIS
gis = GIS("https://pythonapi.playground.esri.com/portal/home")
map = gis.map()
map
```

```python
# Should display correctly with no errors etc
assert map._portal_url == "https://pythonapi.playground.esri.com/portal/"
```

```python
# ASSERT THAT CHANGING THE START/STOP slider changes these values
print(f"start = {map.start_time}, end = {map.end_time}")
```

```python
# ASSERT THAT SETTING THE START/STOP CHANGES THE VALUES
import datetime as dt
map.start_time = dt.datetime(2014, 4, 21)
map.end_time = dt.datetime(2014, 4, 29)
```

```python
# ASSERT THAT THIS MOVES THE WINDOW PROGRAMATICALLY
import time
import datetime as dt
start_date = dt.datetime(2014, 4, 1)
for i in range(1, 30):
    map.start_time = start_date + dt.timedelta(days=i)
    map.end_time = start_date + dt.timedelta(days=i+10)
    time.sleep(0.4)
```

```python
# ASSERT THAT NOW THE ONLY FIRST SLIDER CHANGES THE MAP
map.time_mode = "instance"
```

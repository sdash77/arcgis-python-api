```python
#First test calling map.gallery_basemaps before using portal basemap
from arcgis.gis import GIS
gis = GIS('https://python.playground.esri.com/portal',
          'arcgis_python',
          'amazing_arcgis_123')
map = gis.map()
print(map.gallery_basemaps)
assert 'openstreetmap' in map.gallery_basemaps
map.basemap = 'openstreetmap'
map
```

```python
import arcgis
from arcgis.gis import GIS

gis = GIS("https://www.arcgis.com", "arcgis_python", "amazing_arcgis_123")

m1 = gis.map("San Francisco", zoomlevel=10)
m1
```

```python
x,y = (-13629244.613461372, 4544944.042260825) # any random point
m1.draw({'x':x, 'y':y, 'spatialReference': {"wkid": 3857}}, symbol={"angle":0,"xoffset":0,"yoffset":0,"type":"esriPMS","url":"http://static.arcgis.com/images/Symbols/Basic/RedBeacon.png","contentType":"image/png","width":20,"height":20})
```

```python
m1.clear_graphics()
```

```python
from arcgis.mapping import create_symbol
from arcgis.gis import GIS
from arcgis.geocoding import geocode
gis = GIS()
geocode_fs = geocode(address=None,
                     location=[-118.71511, 34.09042],
                     category="Coffee shop",
                     out_fields="Place_addr, PlaceName",
                     max_locations=25,
                     as_featureset=True)
symbol = create_symbol(geometry_type="point",
                       symbol_style='+') 
g_map = gis.map([34.09042, -118.71511], zoomlevel=11)
g_map.draw(geocode_fs, symbol=symbol)
g_map
```

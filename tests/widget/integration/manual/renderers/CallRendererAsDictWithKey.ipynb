```python
from arcgis.mapping import generate_renderer
from arcgis.gis import GIS
gis = GIS()
layers = gis.content.search("States", "Feature Layer")[0]
layers
m = gis.map("United States")
simple_renderer = generate_renderer(render_type='s',geometry_type='polygon')
layers.layers[0].properties.geometryType
lyr = layers.layers[0]
m.add_layer(item=lyr, options={'renderer': simple_renderer})
m
```

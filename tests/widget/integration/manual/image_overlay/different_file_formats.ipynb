```python
from arcgis.raster import Raster
from arcgis.gis import GIS
gis = GIS()
map = gis.map("Amberg, Germany")
map
```

```python
%%time
map.local_raster_file_format = "png"
assert map.local_raster_file_format == "png"
map.add_layer(Raster(r"./data/090162.tif"))
```

```python
map2 = gis.map("Amberg, Germany")
map2
```

```python
%%time
map2.local_raster_file_format = "jpg"
assert map2.local_raster_file_format == "jpg"
map2.add_layer(Raster(r"./data/090162.tif"))
```

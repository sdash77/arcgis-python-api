```python
from arcgis.gis import GIS
from arcgis.raster import Raster
gis = GIS()
map = gis.map(zoomlevel=5)
map
```

```python
try:
    raster = Raster(r"./data/six_bands.img")
    map.add_layer(raster)
except Exception as e:
    if "subset of bands" in str(e):
        print("Test passed, since this exception about # of bands was raised:\n")
        print(e)
    else:
        raise e
```

```python
raster = Raster(r"./data/six_bands.img")
raster_0 = raster.get_raster_bands()[0]
raster_0.extent = {"xmin":0,
                   "ymin":0,
                   "xmax":15,
                   "ymax":15,
                   "spatialReference": {"wkid": 4326}}
map.add_layer(raster_0)
```

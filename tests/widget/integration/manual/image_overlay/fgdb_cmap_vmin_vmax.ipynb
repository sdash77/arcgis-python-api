```python
import zipfile
zip_ref = zipfile.ZipFile("./data/madison_wi_2.gdb.zip", 'r')
zip_ref.extractall("./data/madison_wi_2.gdb/")
zip_ref.close()

from arcgis.gis import GIS
from arcgis.raster import Raster
gis = GIS()
raster = Raster("./data/madison_wi_2.gdb/temperature_prediction",
                cmap = "OrRd",
                opacity = 0.5)
raster.vmin = 74
raster.vmax = 80
map = gis.map("Madison, WI")
map.add_layer(raster)
map
```

```python
from arcgis.gis import GIS
from arcgis.raster import Raster
gis = GIS()
map = gis.map("Newark, NJ")
raster = Raster("./data/newark_nj_1922.jpg",
                extent = {"xmin":-74.22655,
                          "ymin":40.712216,
                          "xmax":-74.12544,
                          "ymax":40.773941,
                          "spatialReference":{"wkid":4326}},
                opacity = 0.85)
map.add_layer(raster)
map
```

```python
import numpy
import matplotlib.pyplot as plt
from arcgis.gis import GIS
gis = GIS()
map = gis.map()
map
```

```python
from arcgis.raster import Raster
raster = Raster("./data/newark_nj_1922.jpg",
                extent = {"xmin":0,
                          "ymin":0,
                          "xmax": 30,
                          "ymax":30,
                          "spatialReference":{"wkid":4326}},
                opacity = 0.85)
map.add_layer(raster)
```

```python
raster2 = Raster("./data/joined.img")
map.add_layer(raster2)
```

```python
corona_cases = gis.content.get("bbb2e4f589ba40d692fab712ae37b9ac")
map.add_layer(corona_cases)
```

```python
orig_num_layers = len(map.layers)
orig_num_layers
```

```python
# MANUALLY VISUALLY ASSERT THAT BOTH RASTERS ARE REMOVED FROM MAP
map.remove_layers([raster, raster2])
```

```python
assert len(map.layers) == orig_num_layers - 2
len(map.layers)
```

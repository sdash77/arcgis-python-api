```python
import numpy
import matplotlib.pyplot as plt
from arcgis.gis import GIS
map = GIS().map("Newark, NJ")
map
```

```python
from arcgis.raster import Raster
raster = Raster("./data/newark_nj_1922.jpg",
                extent = {"xmin":-74.22655,
                          "ymin":40.712216,
                          "xmax":-74.12544,
                          "ymax":40.773941,
                          "spatialReference":{"wkid":4326}},
                opacity = 0.85)
map.add_layer(raster)
```

```python
assert len(map.layers) == 1
```

```python
# MANUALLY VISUALLY ASSERT THAT RASTER IS REMOVED FROM MAP
map.remove_layers([raster,])
```

```python
assert len(map.layers) == 0
```

```python
from arcgis.gis import GIS
from arcgis.mapping import WebMap

gis = GIS(profile="DavidJVitale")
webmap = WebMap()

webmap
```

```python
# wait for the webmap to load, then run this cell

from arcgis.features import FeatureLayer
hurricane_layer = FeatureLayer(
    url="https://sampleserver6.arcgisonline.com/arcgis/rest/services/Hurricanes/MapServer/0")
webmap.add_layer(hurricane_layer)
```

```python
# wait for the above layer to load, then run this cell
from arcgis.mapping.ogc import *
csv_layer = CSVLayer("https://developers.arcgis.com/javascript/latest/sample-code/layers-csv/live/earthquakes.csv")
webmap.add_layer(csv_layer)
```

```python
# Save the webmap and assert that it looks good in teh webmap viewer
item = webmap.save({'title' : 'test from live webmap',
                    'tags': ['1','2'],
                    'snippet' : "asdfa"})
item
```

```python
# Load the saved webmap portal item and make sure it looks the same
wm2 = WebMap(item)
wm2
```

```python
# Assert that you can add a new layer to the already saved webmap
kml = KMLLayer("https://earthquake.usgs.gov/fdsnws/event/1/query?format=kml&minmagnitude=5.8")
wm2.add_layer(kml)
```

```python
# Make sure all 3 layers are available
item2 = wm2.save({"title" : "Added to existing webmap",
                  "tags" : ['1','2'],
                  "snippet": "123123"})
item2
```

```python
# Assert that adding layers to webmap before displaying still
# shows them in the webmap
wm3 = WebMap()
grss_layer = GeoRSSLayer("https://arcgis.github.io/arcgis-samples-javascript/sample-data/layers-georss/sample-georss.xml")
kml_layer = KMLLayer("https://earthquake.usgs.gov/fdsnws/event/1/query?format=kml&minmagnitude=5.8")
wm3.add_layer(grss_layer)
wm3.add_layer(kml_layer)
wm3
```

```python
wm3.save({"title":'asdfa',
          'tags':['1','2'],
          'snippet':'asdfap'})
```

```python
# Assert that the first layer is removed from th webmap
wm3.remove_layer(wm3.layers[0])
```

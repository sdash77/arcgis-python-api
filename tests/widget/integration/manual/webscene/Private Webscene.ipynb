```python
#FIRST, MAKE A WEBSCENE
from arcgis.gis import GIS
g = GIS(profile="widget_test_publisher_credentials")
webscene_item = g.content.get("421433baeb8d487b903d4a89df79149b")
from arcgis.widgets import MapView
webscene = MapView(gis=g, item=webscene_item, mode="3D")
webscene
```

```python
import datetime
new_webscene_item = webscene.save(({'title':'new zion webscene {}'.format(datetime.datetime.now()),
                             'snippet':'Arterial road network of Italy',
                             'tags':'streets, network, roads'})
)
new_webscene_item
```

```python
#THEN, LOAD THE NEW PRIVATE WEBSCENE
from arcgis.gis import GIS
g2 = GIS(profile="widget_test_publisher_credentials")
from arcgis.mapping import WebScene
webscene = WebScene(g.content.get(new_webscene_item.id))
webscene
```

```python
from arcgis.features import FeatureSet
from arcgis.gis import GIS
with open("./arbitrary_featureset.json", "r") as f:
    fset_json_text = f.read()
fset = FeatureSet.from_json(fset_json_text)
g = GIS()
map = g.map("Pittsburg, PA")
map.draw(fset)
map
```

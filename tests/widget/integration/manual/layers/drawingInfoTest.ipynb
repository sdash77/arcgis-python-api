This method should cause the map widget to automatically refresh and display changes, but won't save them to the webmap unless update() or save() are called.

"online_test" is a profile with arcgis_python geosaurus credentials- recommended to use that account.

```python
import sys

sys.path.insert(0, r"/Users/cowboy/GitHub/np_geo/src")
from arcgis.gis import GIS
from arcgis.mapping import WebMap
```

```python
gis = GIS(profile="online_test", verify_cert = False)
```

```python
# webmap with target layer in a group layer
wm = WebMap(gis.content.get("73e4738d856d4c80bfa2b11f3aeca72b"))
```

The cell below will display the current stream layer (purple lines with black arrows, 50% transparency, black labels with white outline). May need to zoom in to see the labels. At any point, reset back to this default by running the cell.

```python
wm = WebMap(gis.content.get("73e4738d856d4c80bfa2b11f3aeca72b"))
wm
```

The cell below should change the stream to red with green arrows, increase the opacity/decrease transparency, and change the label outline to blue. Feel free to tinker with the settings.

```python
wm.update_drawing_info(layer = test_layer, renderer = test_rend, transparency = 5, label_info = test_label, show_labels = True)
 
```

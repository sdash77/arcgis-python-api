```python
# --------------
# TEST WMS LAYER
# --------------
from arcgis.gis import GIS
from arcgis.mapping.ogc import *
gis = GIS(profile="DavidJVitale")
map1 = gis.map()
wms_layer2 = WMSLayer("http://wms.chartbundle.com/wms")
map1.add_layer(wms_layer2)
map1
```

### Should look like this:
</img>

```python
# ASSERT THAT THE SAVED WEBMAP HAS BOTH OF THE WMS LAYERS ON IT
map1.save({"title": "WMS Layer from MapView",
           "snippet": "Adsf",
           "tags":["1","2"]})
```

```python
# ASSERT LAYER CAN BE REMOVED
map1.remove_layers()
```

```python
# ---------------
# TEST KML LAYER
# ---------------
map3 = gis.map()
k = KMLLayer("https://earthquake.usgs.gov/fdsnws/event/1/query?format=kml&minmagnitude=5.8")
map3.add_layer(k)
map3
```

### Should look like this:

```python
# ASSERT SAVED WEBMAP HAS THE KML LAYER
map3.save({"title": "KML Layer from MapView",
           "snippet": "asdf",
           "tags": ["1","2"]})
```

```python
# ASSERT CAN BE REMOVED
map3.remove_layers()
```

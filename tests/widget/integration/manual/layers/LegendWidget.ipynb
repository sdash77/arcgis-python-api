```python
from arcgis.gis import GIS
gis = GIS()
map = gis.map("Bermuda", zoomlevel=3)
map.legend = True
layer1 = gis.content.search('title:USA freeway system AND owner:esri', 
                                  item_type = 'Feature Layer')[0]
map.add_layer(layer1)
layer2 = gis.content.get('bbf5a6899824407697e114556c213139')
map.add_layer(layer2)

map
```

```python
point_lyrs = []
_3dobject_lyrs = []
intmesh_lyrs = []
pointcloud_lyrs = []
for item in gis.content.search("*", item_type="Scene Service", max_items=999):
    if "Point" in item.typeKeywords:
        point_lyrs.append(item)
    if "3DObject" in item.typeKeywords:
        _3dobject_lyrs.append(item)
    if "IntegratedMesh" in item.typeKeywords:
        intmesh_lyrs.append(item)
    if "PointCloud" in item.typeKeywords:
        pointcloud_lyrs.append(item)
```

```python
from arcgis.gis import GIS
gis = GIS()
agol_map = gis.map("Europe")
agol_map
```

```python
# Wait for this to finish
agol_map.take_screenshot()
```

```python
# Then make sure this assertion works
assert agol_map.print_service_url == 'https://utility.arcgisonline.com/'\
    'arcgis/rest/services/Utilities/PrintingTools/GPServer/'\
    'Export%20Web%20Map%20Task'
```

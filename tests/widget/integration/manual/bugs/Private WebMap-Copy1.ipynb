```python
#FIRST, CREATE THE NEW PRIVATE WEBMAP
from arcgis.gis import *
gis = GIS(profile="widget_test_publisher_credentials")
map1 = gis.map(zoomlevel=10)
map1.center = {'spatialReference': {'latestWkid': 3857, 'wkid': 102100},
 'x': -13163452.471499627,
 'y': 4008459.115353957}
freeway_item = gis.content.get('91c6a5f6410b4991ab0db1d7c26daacb')
map1.add_layer(freeway_item)
freeway_feature_layer = freeway_item.layers[0]
map1.add_layer(freeway_feature_layer, {"renderer":"ClassedSizeRenderer", "field_name": "DIST_MILES"})
map1.basemap = 'national-geographic'
map1
```

```python
import datetime
new_webmap = map1.save({'title' : 'Saved Webmap at {}'.format(datetime.datetime.now()),
           'snippet' : 'test',
           'tags' : 'test'})
new_webmap
```

```python
from arcgis.gis import GIS
g2 = GIS(profile="widget_test_publisher_credentials")
from arcgis.mapping import WebMap
webmap = WebMap(g2.content.get(new_webmap.id))
webmap
```

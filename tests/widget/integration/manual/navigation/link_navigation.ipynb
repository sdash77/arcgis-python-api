```python
from arcgis.gis import GIS
from ipywidgets import HBox
gis = GIS()
map1 = gis.map("Chicago, IL")
map1.basemap = "gray"
map2 = gis.map("Chicago, IL")
map2.basemap = "dark-gray"
map3 = gis.map("Chicago, IL", mode="3D")
map3.basemap = "national-geographic"
map1.sync_navigation(map2)
map2.sync_navigation(map3)
HBox([map1, map2, map3])
```

```python
# ASSERT ALL 3 ARE LINKED TOGETHER PROPERLY
```

```python
# Now, unsync 2<->1. Assert that 1<->3 still works
map2.unsync_navigation(map1)
```

```python
# Now unsync 3<->1, assert that no more syncing takes place
map3.unsync_navigation(map1)
```

```python
# Resync all 3 for next test case
map1.sync_navigation(map2)
map2.sync_navigation(map3)
```

```python
# Assert that 1 is synced to nothing
# Assert that 2<->3 are still synced
map1.unsync_navigation()
```

```python
# Assert that nothing is synced
map2.unsync_navigation(map3)
```

```python
# Resync all 3 for next test case
map1.sync_navigation(map2)
map2.sync_navigation(map3)
```

```python
# Assert that 2 is synced to nothing, but 1<->3 are synced
map2.unsync_navigation()
```

```python
# Unsync everything for next test case
map3.unsync_navigation()
```

```python
# Assert that all 3 maps are synced
map1.sync_navigation([map2, map3])
```

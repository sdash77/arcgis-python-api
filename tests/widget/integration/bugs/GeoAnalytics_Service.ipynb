```python
from arcgis.gis import GIS
import arcgis.geoanalytics
from arcgis.geoanalytics.manage_data import copy_to_data_store
 
portal = "https://gplinux.esri.com/portal"
gis = GIS(portal, "admin", "esri.agp", verify_cert=False)
arcgis.env.default_aggregation_styles = True
bigdata_fileshares = gis.content.search("", item_type = "big data file share")
for layers in bigdata_fileshares:
    if "bigDataFileShares_pytest" in layers.title:
        data_item = layers
 
for lyr in data_item.layers:
    if lyr.properties.name == "ChicagoCrimes":
        layer_to_use = lyr
tool_output = copy_to_data_store(layer_to_use)
self_aggregating_map_service = tool_output.publish()
processed_map  = gis.map("Chicago")
processed_map.add_layer(self_aggregating_map_service)
processed_map
```

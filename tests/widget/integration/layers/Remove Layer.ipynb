```python
#Tests that Item from different portals can be drawn on the same map with add_layer
from arcgis.gis import GIS
g = GIS("https://python.playground.esri.com/portal",
        "arcgis_python",
        "amazing_arcgis_123")
m = g.map("Fresno, CA", 7)
g2 = GIS()
item_from_portal = g.groups.search('LA County Emergency Management')[0].content()[1]
item_from_agol = g2.content.get('bbf5a6899824407697e114556c213139')

layer_portal = item_from_portal.layers[0]
layer_agol = item_from_agol.layers[0]

m.add_layer(layer_portal)
m.add_layer(layer_agol)
m
```

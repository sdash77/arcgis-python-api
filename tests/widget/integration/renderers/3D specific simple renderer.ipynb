```python
from arcgis.gis import GIS
g = GIS()
map = g.map("Angel's Landing UT", 15)
map.mode = "3D"
map.tilt= 40
map.heading = 180
layer = g.content.get('dd0889d7ccd340dd876dac12184e99f9').layers[0]
symbol = {
  "type": "line-3d",  # JS type of new LineSymbol3D()
  "symbolLayers": [{
    "type": "path",  # JS type of new PathSymbol3DLayer()
    "size": 20,  # 20 meters in diameter
    "material": { "color": "#0083ff" }
  }]
}
map.add_layer(layer,  {"renderer" : "autocast",
      "type" : "simple",
      "symbol": symbol}
)
map
```
